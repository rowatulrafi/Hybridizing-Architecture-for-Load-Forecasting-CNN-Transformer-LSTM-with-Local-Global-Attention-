```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torch_geometric.nn import GCNConv
from torch_geometric.data import Data
import optuna
from optuna.trial import TrialState
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error
import warnings
warnings.filterwarnings('ignore')

# Set seeds for reproducibility
SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

# Device configuration
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# Quantile levels for prediction
QUANTILES = [0.1, 0.5, 0.9]

# Hyperparameters (will be tuned by Optuna)
class HyperParams:
    def __init__(self):
        # Default values, will be overridden by Optuna
        self.batch_size = 64
        self.sequence_length = 24  # As specified in the requirements
        self.hidden_size = 128
        self.num_layers = 2
        self.dropout = 0.2
        self.learning_rate = 0.001
        self.weight_decay = 1e-5
        self.cnn_filters = 64
        self.cnn_kernel_size = 3
        self.nhead = 8  # Number of heads in transformer
        self.transformer_dim = 128
        self.transformer_layers = 2
        self.gnn_hidden = 64
        self.patience = 10  # Early stopping patience

# Custom dataset class
class ElectricityLoadDataset(Dataset):
    def __init__(self, features, targets, sequence_length):
        self.features = features
        self.targets = targets
        self.sequence_length = sequence_length

    def __len__(self):
        return len(self.features) - self.sequence_length

    def __getitem__(self, idx):
        x = self.features[idx:idx + self.sequence_length]
        y = self.targets[idx + self.sequence_length]
        return torch.FloatTensor(x), torch.FloatTensor([y])

# CNN Block
class CNNBlock(nn.Module):
    def __init__(self, in_channels, out_channels, kernel_size, dropout):
        super(CNNBlock, self).__init__()
        self.conv1 = nn.Conv1d(in_channels, out_channels, kernel_size, padding=kernel_size//2)
        self.bn1 = nn.BatchNorm1d(out_channels)
        self.conv2 = nn.Conv1d(out_channels, out_channels, kernel_size, padding=kernel_size//2)
        self.bn2 = nn.BatchNorm1d(out_channels)
        self.relu = nn.ReLU()
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # x shape: [batch, sequence, features]
        x = x.permute(0, 2, 1)  # Change to [batch, features, sequence]
        x = self.relu(self.bn1(self.conv1(x)))
        x = self.dropout(x)
        x = self.relu(self.bn2(self.conv2(x)))
        x = self.dropout(x)
        x = x.permute(0, 2, 1)  # Change back to [batch, sequence, features]
        return x

# Transformer Block
class TransformerBlock(nn.Module):
    def __init__(self, d_model, nhead, dim_feedforward, dropout, num_layers):
        super(TransformerBlock, self).__init__()
        encoder_layers = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=nhead,
            dim_feedforward=dim_feedforward,
            dropout=dropout,
            batch_first=True  # For [batch, seq, feature] input
        )
        self.transformer_encoder = nn.TransformerEncoder(encoder_layers, num_layers=num_layers)
        self.pos_encoder = PositionalEncoding(d_model, dropout)

    def forward(self, x):
        # Add positional encoding
        x = self.pos_encoder(x)
        # Create a mask for padding if needed
        # mask = self._generate_square_subsequent_mask(x.size(1)).to(x.device)
        # Apply transformer
        return self.transformer_encoder(x)

    def _generate_square_subsequent_mask(self, sz):
        mask = (torch.triu(torch.ones(sz, sz)) == 1).transpose(0, 1)
        mask = mask.float().masked_fill(mask == 0, float('-inf')).masked_fill(mask == 1, float(0.0))
        return mask

# Positional Encoding for Transformer
class PositionalEncoding(nn.Module):
    def __init__(self, d_model, dropout=0.1, max_len=5000):
        super(PositionalEncoding, self).__init__()
        self.dropout = nn.Dropout(p=dropout)

        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-np.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        pe = pe.unsqueeze(0)
        self.register_buffer('pe', pe)

    def forward(self, x):
        x = x + self.pe[:, :x.size(1), :]
        return self.dropout(x)

# GNN Block
class GNNBlock(nn.Module):
    def __init__(self, in_channels, hidden_channels):
        super(GNNBlock, self).__init__()
        self.conv1 = GCNConv(in_channels, hidden_channels)
        self.conv2 = GCNConv(hidden_channels, hidden_channels)
        self.relu = nn.ReLU()

    def forward(self, x, edge_index, edge_weight=None):
        x = self.relu(self.conv1(x, edge_index, edge_weight))
        x = self.conv2(x, edge_index, edge_weight)
        return x

# Combined CNN-Transformer-LSTM-GNN Model
class CNNTransformerLSTMGNN(nn.Module):
    def __init__(self, input_dim, output_dim, params):
        super(CNNTransformerLSTMGNN, self).__init__()

        self.params = params

        # CNN to extract local patterns
        self.cnn = CNNBlock(
            in_channels=input_dim,
            out_channels=params.cnn_filters,
            kernel_size=params.cnn_kernel_size,
            dropout=params.dropout
        )

        # Linear layer to map CNN output to transformer dimension
        self.cnn_to_transformer = nn.Linear(params.cnn_filters, params.transformer_dim)

        # Transformer for capturing global dependencies
        self.transformer = TransformerBlock(
            d_model=params.transformer_dim,
            nhead=params.nhead,
            dim_feedforward=params.transformer_dim * 4,
            dropout=params.dropout,
            num_layers=params.transformer_layers
        )

        # LSTM for temporal dependencies
        self.lstm = nn.LSTM(
            input_size=params.transformer_dim,
            hidden_size=params.hidden_size,
            num_layers=params.num_layers,
            batch_first=True,
            dropout=params.dropout if params.num_layers > 1 else 0,
            bidirectional=True
        )

        # GNN for spatial dependencies
        self.gnn = GNNBlock(
            in_channels=params.hidden_size * 2,  # * 2 for bidirectional LSTM
            hidden_channels=params.gnn_hidden
        )

        # Attention mechanism
        self.attention = nn.MultiheadAttention(
            embed_dim=params.hidden_size * 2,
            num_heads=params.nhead,
            dropout=params.dropout,
            batch_first=True
        )

        # Output layers for each quantile
        self.output_layers = nn.ModuleList([
            nn.Linear(params.hidden_size * 2 + params.gnn_hidden, output_dim)
            for _ in QUANTILES
        ])

    def forward(self, x, edge_index=None, edge_weight=None):
        batch_size, seq_len, features = x.size()

        # Apply CNN
        x_cnn = self.cnn(x)

        # Map to transformer dimension
        x_transformer_input = self.cnn_to_transformer(x_cnn)

        # Apply Transformer
        x_transformer = self.transformer(x_transformer_input)

        # Apply LSTM
        x_lstm, (h_n, c_n) = self.lstm(x_transformer)

        # Create GNN graph if edge_index is None
        if edge_index is None:
            # Create a simple temporal graph where each node connects to its neighbors
            node_indices = torch.arange(seq_len, device=x.device)
            source_nodes = node_indices[:-1].repeat(2)
            target_nodes = torch.cat([node_indices[1:], node_indices[:-1]])
            edge_index = torch.stack([source_nodes, target_nodes], dim=0)

            # Create edge weights based on proximity
            if edge_weight is None:
                edge_weight = torch.ones(edge_index.size(1), device=x.device)

        # Reshape for GNN (treat each time step as a node)
        x_gnn_input = x_lstm.reshape(-1, x_lstm.size(-1))  # [batch * seq, hidden]

        # Clone edge_index for each batch
        batch_edge_indices = []
        batch_edge_weights = []
        for i in range(batch_size):
            offset = i * seq_len
            batch_edge_index = edge_index.clone()
            batch_edge_index += offset
            batch_edge_indices.append(batch_edge_index)

            if edge_weight is not None:
                batch_edge_weights.append(edge_weight.clone())

        # Concatenate edge indices and weights
        batch_edge_index = torch.cat(batch_edge_indices, dim=1)
        if edge_weight is not None:
            batch_edge_weight = torch.cat(batch_edge_weights, dim=0)
        else:
            batch_edge_weight = None

        # Apply GNN
        x_gnn = self.gnn(x_gnn_input, batch_edge_index, batch_edge_weight)

        # Reshape back to [batch, seq, hidden]
        x_gnn = x_gnn.reshape(batch_size, seq_len, -1)

        # Apply attention over the LSTM outputs
        x_attn, _ = self.attention(x_lstm, x_lstm, x_lstm)

        # Concatenate GNN and Attention outputs
        x_combined = torch.cat([x_attn, x_gnn], dim=-1)

        # Get the last time step for prediction
        x_last = x_combined[:, -1, :]

        # Apply output layers for each quantile
        quantile_outputs = [output_layer(x_last) for output_layer in self.output_layers]

        # Stack quantile outputs
        return torch.stack(quantile_outputs, dim=1)

# Pinball Loss for Quantile Regression
def pinball_loss(y_pred, y_true, quantiles=QUANTILES):
    losses = []
    for i, q in enumerate(quantiles):
        errors = y_true - y_pred[:, i]
        losses.append(torch.max(q * errors, (q - 1) * errors))

    return torch.mean(torch.stack(losses))

# Function to preprocess data
def preprocess_data(df):
    # Convert datetime to index
    df['datetime'] = pd.to_datetime(df['datetime'])
    df.set_index('datetime', inplace=True)

    # Extract year for train/val/test split
    df['year'] = df.index.year

    # Handle missing values (if any)
    df.fillna(method='ffill', inplace=True)
    df.fillna(method='bfill', inplace=True)  # In case there are missing values at the beginning

    # Extract features and target
    target_column = 'nat_demand'
    feature_columns = [col for col in df.columns if col != target_column and col != 'year']

    # Create time features
    df['hour'] = df.index.hour
    df['month'] = df.index.month
    df['day'] = df.index.day

    # Add sine and cosine transformations for cyclical features
    for col in ['hour', 'month', 'day']:
        if col == 'hour':
            max_val = 24
        elif col == 'month':
            max_val = 12
        else:  # day
            max_val = 31

        df[f'{col}_sin'] = np.sin(2 * np.pi * df[col] / max_val)
        df[f'{col}_cos'] = np.cos(2 * np.pi * df[col] / max_val)

        # Remove original columns
        feature_columns.remove(col) if col in feature_columns else None

    # Add cyclical features to feature columns
    for col in ['hour', 'month', 'day']:
        feature_columns.extend([f'{col}_sin', f'{col}_cos'])

    # Create lags for the target variable
    for lag in [1, 2, 3, 6, 12, 24, 48, 72, 168]:  # 1h, 2h, 3h, 6h, 12h, 1d, 2d, 3d, 1w
        df[f'nat_demand_lag_{lag}'] = df[target_column].shift(lag)
        feature_columns.append(f'nat_demand_lag_{lag}')

    # Create rolling statistics
    for window in [6, 12, 24, 48, 72, 168]:  # 6h, 12h, 1d, 2d, 3d, 1w
        df[f'nat_demand_rolling_mean_{window}'] = df[target_column].rolling(window=window).mean().shift(1)
        df[f'nat_demand_rolling_std_{window}'] = df[target_column].rolling(window=window).std().shift(1)
        df[f'nat_demand_rolling_max_{window}'] = df[target_column].rolling(window=window).max().shift(1)
        df[f'nat_demand_rolling_min_{window}'] = df[target_column].rolling(window=window).min().shift(1)

        feature_columns.extend([
            f'nat_demand_rolling_mean_{window}',
            f'nat_demand_rolling_std_{window}',
            f'nat_demand_rolling_max_{window}',
            f'nat_demand_rolling_min_{window}'
        ])

    # Drop rows with NaN values (due to lagging and rolling operations)
    df.dropna(inplace=True)

    # Split data based on year
    train_data = df[df['year'].isin([2015, 2016, 2017])]
    val_data = df[df['year'] == 2018]
    test_data = df[~df['year'].isin([2015, 2016, 2017, 2018])]

    # Remove 'year' column
    train_data = train_data.drop('year', axis=1)
    val_data = val_data.drop('year', axis=1)
    test_data = test_data.drop('year', axis=1)

    # Scale features
    scaler_X = StandardScaler()
    scaler_y = StandardScaler()

    # Fit on training data
    train_features = scaler_X.fit_transform(train_data[feature_columns])
    val_features = scaler_X.transform(val_data[feature_columns])
    test_features = scaler_X.transform(test_data[feature_columns])

    # Scale target separately
    train_target = scaler_y.fit_transform(train_data[[target_column]])
    val_target = scaler_y.transform(val_data[[target_column]])
    test_target = scaler_y.transform(test_data[[target_column]])

    # Flatten target arrays
    train_target = train_target.flatten()
    val_target = val_target.flatten()
    test_target = test_target.flatten()

    # Create datasets
    return {
        'train_features': train_features,
        'train_target': train_target,
        'val_features': val_features,
        'val_target': val_target,
        'test_features': test_features,
        'test_target': test_target,
        'feature_columns': feature_columns,
        'target_column': target_column,
        'scaler_X': scaler_X,
        'scaler_y': scaler_y,
        'train_data': train_data,
        'val_data': val_data,
        'test_data': test_data
    }

# Create data loaders
def create_dataloaders(data_dict, params):
    # Create datasets
    train_dataset = ElectricityLoadDataset(
        data_dict['train_features'],
        data_dict['train_target'],
        params.sequence_length
    )

    val_dataset = ElectricityLoadDataset(
        data_dict['val_features'],
        data_dict['val_target'],
        params.sequence_length
    )

    test_dataset = ElectricityLoadDataset(
        data_dict['test_features'],
        data_dict['test_target'],
        params.sequence_length
    )

    # Create data loaders
    train_loader = DataLoader(
        train_dataset,
        batch_size=params.batch_size,
        shuffle=True,
        num_workers=4,
        pin_memory=True
    )

    val_loader = DataLoader(
        val_dataset,
        batch_size=params.batch_size,
        shuffle=False,
        num_workers=4,
        pin_memory=True
    )

    test_loader = DataLoader(
        test_dataset,
        batch_size=params.batch_size,
        shuffle=False,
        num_workers=4,
        pin_memory=True
    )

    return train_loader, val_loader, test_loader

# Define the evaluation metrics
def calculate_metrics(y_true, y_pred):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)

    # MSE
    mse = mean_squared_error(y_true, y_pred)
    # RMSE
    rmse = np.sqrt(mse)
    # MAE
    mae = mean_absolute_error(y_true, y_pred)
    # NRMSE (Normalized RMSE)
    nrmse = rmse / (np.max(y_true) - np.min(y_true))
    # MAPE (Mean Absolute Percentage Error)
    mape = np.mean(np.abs((y_true - y_pred) / y_true)) * 100

    return {
        'MSE': mse,
        'RMSE': rmse,
        'MAE': mae,
        'NRMSE': nrmse,
        'MAPE': mape
    }

# Training function
def train(model, train_loader, val_loader, optimizer, scheduler, num_epochs, patience, device):
    train_losses = []
    val_losses = []
    best_val_loss = float('inf')
    patience_counter = 0
    best_model = None

    for epoch in range(num_epochs):
        # Training
        model.train()
        train_loss = 0

        for X_batch, y_batch in train_loader:
            X_batch, y_batch = X_batch.to(device), y_batch.to(device)

            # Forward pass
            y_pred = model(X_batch)

            # Calculate loss
            loss = pinball_loss(y_pred, y_batch)

            # Backward and optimize
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

            train_loss += loss.item()

        train_loss /= len(train_loader)
        train_losses.append(train_loss)

        # Validation
        model.eval()
        val_loss = 0

        with torch.no_grad():
            for X_batch, y_batch in val_loader:
                X_batch, y_batch = X_batch.to(device), y_batch.to(device)

                # Forward pass
                y_pred = model(X_batch)

                # Calculate loss
                loss = pinball_loss(y_pred, y_batch)
                val_loss += loss.item()

        val_loss /= len(val_loader)
        val_losses.append(val_loss)

        # Update learning rate
        scheduler.step(val_loss)

        # Print progress
        print(f'Epoch [{epoch+1}/{num_epochs}], Train Loss: {train_loss:.4f}, Val Loss: {val_loss:.4f}')

        # Check for early stopping
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            patience_counter = 0
            # Save best model
            best_model = model.state_dict().copy()
        else:
            patience_counter += 1
            if patience_counter >= patience:
                print(f'Early stopping at epoch {epoch+1}')
                break

    # Load best model
    model.load_state_dict(best_model)

    return model, train_losses, val_losses

# Evaluation function
def evaluate(model, test_loader, scaler_y, device):
    model.eval()

    all_preds = []
    all_targets = []

    with torch.no_grad():
        for X_batch, y_batch in test_loader:
            X_batch, y_batch = X_batch.to(device), y_batch.to(device)

            # Forward pass
            y_pred = model(X_batch)

            # Extract median prediction (quantile 0.5)
            median_idx = QUANTILES.index(0.5)
            y_pred_median = y_pred[:, median_idx]

            # Convert to numpy and transform back
            y_pred_np = y_pred_median.cpu().numpy()
            y_batch_np = y_batch.cpu().numpy()

            # Store predictions and targets
            all_preds.extend(y_pred_np)
            all_targets.extend(y_batch_np)

    # Convert to numpy arrays
    all_preds = np.array(all_preds).reshape(-1, 1)
    all_targets = np.array(all_targets).reshape(-1, 1)

    # Inverse transform to original scale
    all_preds = scaler_y.inverse_transform(all_preds)
    all_targets = scaler_y.inverse_transform(all_targets)

    # Calculate metrics
    metrics = calculate_metrics(all_targets, all_preds)

    return all_preds, all_targets, metrics

# Plot training and validation loss
def plot_losses(train_losses, val_losses):
    plt.figure(figsize=(10, 6))
    plt.plot(train_losses, label='Train Loss')
    plt.plot(val_losses, label='Validation Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.title('Training and Validation Loss')
    plt.legend()
    plt.grid(True)
    plt.savefig('loss_plot.png')
    plt.show()

# Plot predictions vs actual
def plot_predictions(y_true, y_pred, title='Predictions vs Actual'):
    plt.figure(figsize=(15, 8))
    plt.plot(y_true, label='Actual', color='blue')
    plt.plot(y_pred, label='Predicted', color='red', alpha=0.7)
    plt.xlabel('Time Step')
    plt.ylabel('Electricity Load (MWh)')
    plt.title(title)
    plt.legend()
    plt.grid(True)
    plt.savefig('predictions_plot.png')
    plt.show()

# Optuna objective function for hyperparameter tuning
def objective(trial):
    # Define the hyperparameters to optimize
    params = HyperParams()
    params.batch_size = trial.suggest_categorical('batch_size', [32, 64, 128])
    params.hidden_size = trial.suggest_categorical('hidden_size', [64, 128, 256])
    params.num_layers = trial.suggest_int('num_layers', 1, 3)
    params.dropout = trial.suggest_float('dropout', 0.1, 0.5)
    params.learning_rate = trial.suggest_loguniform('learning_rate', 1e-4, 1e-2)
    params.weight_decay = trial.suggest_loguniform('weight_decay', 1e-6, 1e-4)
    params.cnn_filters = trial.suggest_categorical('cnn_filters', [32, 64, 128])
    params.cnn_kernel_size = trial.suggest_categorical('cnn_kernel_size', [3, 5, 7])
    params.nhead = trial.suggest_categorical('nhead', [4, 8])
    params.transformer_dim = trial.suggest_categorical('transformer_dim', [64, 128, 256])
    params.transformer_layers = trial.suggest_int('transformer_layers', 1, 3)
    params.gnn_hidden = trial.suggest_categorical('gnn_hidden', [32, 64, 128])

    # Create data loaders
    train_loader, val_loader, _ = create_dataloaders(data_dict, params)

    # Create model
    input_dim = data_dict['train_features'].shape[1]
    output_dim = 1  # Predicting a single value

    model = CNNTransformerLSTMGNN(input_dim, output_dim, params).to(device)

    # Define optimizer and scheduler
    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=params.learning_rate,
        weight_decay=params.weight_decay
    )

    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode='min',
        factor=0.5,
        patience=5,
        verbose=True
    )

    # Train for fewer epochs during optimization
    num_epochs = 30
    patience = 7

    # Train model
    _, train_losses, val_losses = train(
        model,
        train_loader,
        val_loader,
        optimizer,
        scheduler,
        num_epochs,
        patience,
        device
    )

    # Return the best validation loss
    return min(val_losses)

# Main function
def main():
    # Load and preprocess data
    print("Loading and preprocessing data...")

    # Sample code to load data (replace with your actual data loading)
    df = pd.read_csv('/content/drive/MyDrive/thesis Final/Implementation/Panama Demand.csv')
    # global data_dict
    global data_dict
    data_dict = preprocess_data(df)

    # Run Optuna optimization
    print("Running hyperparameter optimization with Optuna...")
    study = optuna.create_study(direction='minimize')
    study.optimize(objective, n_trials=20)

    # Get the best hyperparameters
    best_params = study.best_params
    print(f"Best hyperparameters: {best_params}")

    # Apply best hyperparameters
    params = HyperParams()
    params.batch_size = best_params['batch_size']
    params.hidden_size = best_params['hidden_size']
    params.num_layers = best_params['num_layers']
    params.dropout = best_params['dropout']
    params.learning_rate = best_params['learning_rate']
    params.weight_decay = best_params['weight_decay']
    params.cnn_filters = best_params['cnn_filters']
    params.cnn_kernel_size = best_params['cnn_kernel_size']
    params.nhead = best_params['nhead']
    params.transformer_dim = best_params['transformer_dim']
    params.transformer_layers = best_params['transformer_layers']
    params.gnn_hidden = best_params['gnn_hidden']

    # Create data loaders with best parameters
    train_loader, val_loader, test_loader = create_dataloaders(data_dict, params)

    # Create model with best parameters
    input_dim = data_dict['train_features'].shape[1]
    output_dim = 1  # Predicting a single value

    model = CNNTransformerLSTMGNN(input_dim, output_dim, params).to(device)

    # Define optimizer and scheduler
    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=params.learning_rate,
        weight_decay=params.weight_decay
    )

    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode='min',
        factor=0.5,
        patience=10,
        verbose=True
    )

    # Train the final model
    print("Training final model with best hyperparameters...")
    num_epochs = 100  # More epochs for final training
    patience = params.patience  # Early stopping patience

    final_model, train_losses, val_losses = train(
        model,
        train_loader,
        val_loader,
        optimizer,
        scheduler,
        num_epochs,
        patience,
        device
    )

    # Plot training and validation losses
    plot_losses(train_losses, val_losses)

    # Evaluate on test set
    print("Evaluating model on test set...")
    predictions, targets, metrics = evaluate(final_model, test_loader, data_dict['scaler_y'], device)

    # Print metrics
    print("\nTest Set Metrics:")
    for metric_name, metric_value in metrics.items():
        print(f"{metric_name}: {metric_value:.4f}")

    # Plot predictions vs actual
    print("Plotting predictions vs actual...")
    plot_predictions(targets.flatten(), predictions.flatten(), title='CNN-Transformer-LSTM-GNN: Predictions vs Actual')

    # Plot a section of the predictions for better visibility
    # Take a smaller window (e.g., 2 weeks) for more detailed visualization
    window_size = 24 * 14  # 14 days assuming hourly data
    if len(targets) > window_size:
        start_idx = len(targets) // 2 - window_size // 2  # Center the window
        end_idx = start_idx + window_size

        # Plot the zoomed section
        plt.figure(figsize=(15, 8))
        plt.plot(targets[start_idx:end_idx], label='Actual', color='blue')
        plt.plot(predictions[start_idx:end_idx], label='Predicted', color='red', alpha=0.7)
        plt.xlabel('Time Step')
        plt.ylabel('Electricity Load (MWh)')
        plt.title('CNN-Transformer-LSTM-GNN: Predictions vs Actual (2-Week Window)')
        plt.legend()
        plt.grid(True)
        plt.savefig('zoomed_predictions_plot.png')
        plt.show()

    # Plot prediction intervals (10th, 50th, and 90th percentiles)
    print("Plotting prediction intervals...")
    model.eval()

    # Select a sample of the test data
    sample_size = min(window_size, len(data_dict['test_target']) - params.sequence_length)
    sample_start = (len(data_dict['test_target']) - params.sequence_length - sample_size) // 2

    # Prepare sample data
    sample_X = torch.FloatTensor(data_dict['test_features'][sample_start:sample_start+sample_size+params.sequence_length])
    sample_y = torch.FloatTensor(data_dict['test_target'][sample_start:sample_start+sample_size+params.sequence_length])

    # Create sequences
    X_sequences = []
    y_values = []
    for i in range(sample_size):
        X_sequences.append(sample_X[i:i+params.sequence_length])
        y_values.append(sample_y[i+params.sequence_length])

    X_tensor = torch.stack(X_sequences).to(device)

    # Get predictions for all quantiles
    with torch.no_grad():
        all_quantile_preds = model(X_tensor).cpu().numpy()

    # Convert predictions and actual values back to original scale
    quantile_preds = {}
    for i, q in enumerate(QUANTILES):
        preds = all_quantile_preds[:, i, :].reshape(-1, 1)
        quantile_preds[q] = data_dict['scaler_y'].inverse_transform(preds).flatten()

    actual = data_dict['scaler_y'].inverse_transform(np.array(y_values).reshape(-1, 1)).flatten()

    # Plot
    plt.figure(figsize=(15, 8))
    plt.plot(actual, label='Actual', color='black', linewidth=2)
    plt.plot(quantile_preds[0.5], label='Median (50th percentile)', color='red', linewidth=1.5)

    # Create and plot the prediction interval
    plt.fill_between(
        range(len(actual)),
        quantile_preds[0.1],
        quantile_preds[0.9],
        color='red',
        alpha=0.2,
        label='80% Prediction Interval (10th-90th percentile)'
    )

    plt.xlabel('Time Step')
    plt.ylabel('Electricity Load (MWh)')
    plt.title('CNN-Transformer-LSTM-GNN: Prediction Intervals')
    plt.legend()
    plt.grid(True)
    plt.savefig('prediction_intervals_plot.png')
    plt.show()

    # Save the model
    torch.save(final_model.state_dict(), 'cnn_transformer_lstm_gnn_model.pth')
    print("Model saved successfully!")

    return final_model, metrics



# Example of how to run the entire pipeline
if __name__ == "__main__":
    # Replace with your actual data loading code
    df = pd.read_csv('/content/drive/MyDrive/thesis Final/Implementation/Panama Demand.csv')

    # Process the data
    data_dict = preprocess_data(df)

    # Run the main function
    model, metrics = main()

    print("\nFinal Test Set Metrics:")
    for metric_name, metric_value in metrics.items():
        print(f"{metric_name}: {metric_value:.4f}")
```
