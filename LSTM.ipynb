```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
import optuna
from optuna.trial import TrialState
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error
import warnings
warnings.filterwarnings('ignore')

# Set seeds for reproducibility
SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

# Device configuration
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# Quantile levels for prediction
QUANTILES = [0.1, 0.5, 0.9]

# Hyperparameters (will be tuned by Optuna)
class HyperParams:
    def __init__(self):
        # Default values, will be overridden by Optuna
        self.batch_size = 64
        self.sequence_length = 24  # As specified in the requirements
        self.hidden_size = 128
        self.num_layers = 2
        self.dropout = 0.2
        self.learning_rate = 0.001
        self.weight_decay = 1e-5
        self.patience = 10  # Early stopping patience

# Custom dataset class
class ElectricityLoadDataset(Dataset):
    def __init__(self, features, targets, sequence_length):
        self.features = features
        self.targets = targets
        self.sequence_length = sequence_length

    def __len__(self):
        return len(self.features) - self.sequence_length

    def __getitem__(self, idx):
        x = self.features[idx:idx + self.sequence_length]
        y = self.targets[idx + self.sequence_length]
        return torch.FloatTensor(x), torch.FloatTensor([y])

# LSTM-only Model
class LSTM(nn.Module):
    def __init__(self, input_dim, output_dim, params):
        super(LSTM, self).__init__()

        self.params = params

        # LSTM layer
        self.lstm = nn.LSTM(
            input_size=input_dim,
            hidden_size=params.hidden_size,
            num_layers=params.num_layers,
            batch_first=True,
            dropout=params.dropout if params.num_layers > 1 else 0,
            bidirectional=True
        )

        # Dropout layer
        self.dropout = nn.Dropout(params.dropout)

        # Output layers for each quantile
        self.output_layers = nn.ModuleList([
            nn.Linear(params.hidden_size * 2, output_dim)  # * 2 for bidirectional
            for _ in QUANTILES
        ])

    def forward(self, x):
        # Apply LSTM
        x_lstm, (h_n, c_n) = self.lstm(x)

        # Get the output from the last time step
        x_last = x_lstm[:, -1, :]

        # Apply dropout
        x_last = self.dropout(x_last)

        # Apply output layers for each quantile
        quantile_outputs = [output_layer(x_last) for output_layer in self.output_layers]

        # Stack quantile outputs
        return torch.stack(quantile_outputs, dim=1)

# Pinball Loss for Quantile Regression
def pinball_loss(y_pred, y_true, quantiles=QUANTILES):
    losses = []
    for i, q in enumerate(quantiles):
        errors = y_true - y_pred[:, i]
        losses.append(torch.max(q * errors, (q - 1) * errors))

    return torch.mean(torch.stack(losses))

# Function to preprocess data
def preprocess_data(df):
    # Convert datetime to index
    df['datetime'] = pd.to_datetime(df['datetime'])
    df.set_index('datetime', inplace=True)

    # Extract year for train/val/test split
    df['year'] = df.index.year

    # Handle missing values (if any)
    df.fillna(method='ffill', inplace=True)
    df.fillna(method='bfill', inplace=True)  # In case there are missing values at the beginning

    # Extract features and target
    target_column = 'nat_demand'
    feature_columns = [col for col in df.columns if col != target_column and col != 'year']

    # Create time features
    df['hour'] = df.index.hour
    df['month'] = df.index.month
    df['day'] = df.index.day

    # Add sine and cosine transformations for cyclical features
    for col in ['hour', 'month', 'day']:
        if col == 'hour':
            max_val = 24
        elif col == 'month':
            max_val = 12
        else:  # day
            max_val = 31

        df[f'{col}_sin'] = np.sin(2 * np.pi * df[col] / max_val)
        df[f'{col}_cos'] = np.cos(2 * np.pi * df[col] / max_val)

        # Remove original columns
        feature_columns.remove(col) if col in feature_columns else None

    # Add cyclical features to feature columns
    for col in ['hour', 'month', 'day']:
        feature_columns.extend([f'{col}_sin', f'{col}_cos'])

    # Create lags for the target variable
    for lag in [1, 2, 3, 6, 12, 24, 48, 72, 168]:  # 1h, 2h, 3h, 6h, 12h, 1d, 2d, 3d, 1w
        df[f'nat_demand_lag_{lag}'] = df[target_column].shift(lag)
        feature_columns.append(f'nat_demand_lag_{lag}')

    # Create rolling statistics
    for window in [6, 12, 24, 48, 72, 168]:  # 6h, 12h, 1d, 2d, 3d, 1w
        df[f'nat_demand_rolling_mean_{window}'] = df[target_column].rolling(window=window).mean().shift(1)
        df[f'nat_demand_rolling_std_{window}'] = df[target_column].rolling(window=window).std().shift(1)
        df[f'nat_demand_rolling_max_{window}'] = df[target_column].rolling(window=window).max().shift(1)
        df[f'nat_demand_rolling_min_{window}'] = df[target_column].rolling(window=window).min().shift(1)

        feature_columns.extend([
            f'nat_demand_rolling_mean_{window}',
            f'nat_demand_rolling_std_{window}',
            f'nat_demand_rolling_max_{window}',
            f'nat_demand_rolling_min_{window}'
        ])

    # Drop rows with NaN values (due to lagging and rolling operations)
    df.dropna(inplace=True)

    # Split data based on year
    train_data = df[df['year'].isin([2015, 2016, 2017])]
    val_data = df[df['year'] == 2018]
    test_data = df[~df['year'].isin([2015, 2016, 2017, 2018])]

    # Remove 'year' column
    train_data = train_data.drop('year', axis=1)
    val_data = val_data.drop('year', axis=1)
    test_data = test_data.drop('year', axis=1)

    # Scale features
    scaler_X = StandardScaler()
    scaler_y = StandardScaler()

    # Fit on training data
    train_features = scaler_X.fit_transform(train_data[feature_columns])
    val_features = scaler_X.transform(val_data[feature_columns])
    test_features = scaler_X.transform(test_data[feature_columns])

    # Scale target separately
    train_target = scaler_y.fit_transform(train_data[[target_column]])
    val_target = scaler_y.transform(val_data[[target_column]])
    test_target = scaler_y.transform(test_data[[target_column]])

    # Flatten target arrays
    train_target = train_target.flatten()
    val_target = val_target.flatten()
    test_target = test_target.flatten()

    # Create datasets
    return {
        'train_features': train_features,
        'train_target': train_target,
        'val_features': val_features,
        'val_target': val_target,
        'test_features': test_features,
        'test_target': test_target,
        'feature_columns': feature_columns,
        'target_column': target_column,
        'scaler_X': scaler_X,
        'scaler_y': scaler_y,
        'train_data': train_data,
        'val_data': val_data,
        'test_data': test_data
    }

# Create data loaders
def create_dataloaders(data_dict, params):
    # Create datasets
    train_dataset = ElectricityLoadDataset(
        data_dict['train_features'],
        data_dict['train_target'],
        params.sequence_length
    )

    val_dataset = ElectricityLoadDataset(
        data_dict['val_features'],
        data_dict['val_target'],
        params.sequence_length
    )

    test_dataset = ElectricityLoadDataset(
        data_dict['test_features'],
        data_dict['test_target'],
        params.sequence_length
    )

    # Create data loaders
    train_loader = DataLoader(
        train_dataset,
        batch_size=params.batch_size,
        shuffle=True,
        num_workers=4,
        pin_memory=True
    )

    val_loader = DataLoader(
        val_dataset,
        batch_size=params.batch_size,
        shuffle=False,
        num_workers=4,
        pin_memory=True
    )

    test_loader = DataLoader(
        test_dataset,
        batch_size=params.batch_size,
        shuffle=False,
        num_workers=4,
        pin_memory=True
    )

    return train_loader, val_loader, test_loader

# Define the evaluation metrics
def calculate_metrics(y_true, y_pred):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)

    # MSE
    mse = mean_squared_error(y_true, y_pred)
    # RMSE
    rmse = np.sqrt(mse)
    # MAE
    mae = mean_absolute_error(y_true, y_pred)
    # NRMSE (Normalized RMSE)
    nrmse = rmse / (np.max(y_true) - np.min(y_true))
    # MAPE (Mean Absolute Percentage Error)
    mape = np.mean(np.abs((y_true - y_pred) / y_true)) * 100

    return {
        'MSE': mse,
        'RMSE': rmse,
        'MAE': mae,
        'NRMSE': nrmse,
        'MAPE': mape
    }

# Training function
def train(model, train_loader, val_loader, optimizer, scheduler, num_epochs, patience, device):
    train_losses = []
    val_losses = []
    best_val_loss = float('inf')
    patience_counter = 0
    best_model = None

    for epoch in range(num_epochs):
        # Training
        model.train()
        train_loss = 0

        for X_batch, y_batch in train_loader:
            X_batch, y_batch = X_batch.to(device), y_batch.to(device)

            # Forward pass
            y_pred = model(X_batch)

            # Calculate loss
            loss = pinball_loss(y_pred, y_batch)

            # Backward and optimize
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

            train_loss += loss.item()

        train_loss /= len(train_loader)
        train_losses.append(train_loss)

        # Validation
        model.eval()
        val_loss = 0

        with torch.no_grad():
            for X_batch, y_batch in val_loader:
                X_batch, y_batch = X_batch.to(device), y_batch.to(device)

                # Forward pass
                y_pred = model(X_batch)

                # Calculate loss
                loss = pinball_loss(y_pred, y_batch)
                val_loss += loss.item()

        val_loss /= len(val_loader)
        val_losses.append(val_loss)

        # Update learning rate
        scheduler.step(val_loss)

        # Print progress
        print(f'Epoch [{epoch+1}/{num_epochs}], Train Loss: {train_loss:.4f}, Val Loss: {val_loss:.4f}')

        # Check for early stopping
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            patience_counter = 0
            # Save best model
            best_model = model.state_dict().copy()
        else:
            patience_counter += 1
            if patience_counter >= patience:
                print(f'Early stopping at epoch {epoch+1}')
                break

    # Load best model
    model.load_state_dict(best_model)

    return model, train_losses, val_losses

# Evaluation function
def evaluate(model, test_loader, scaler_y, device):
    model.eval()

    all_preds = []
    all_targets = []

    with torch.no_grad():
        for X_batch, y_batch in test_loader:
            X_batch, y_batch = X_batch.to(device), y_batch.to(device)

            # Forward pass
            y_pred = model(X_batch)

            # Extract median prediction (quantile 0.5)
            median_idx = QUANTILES.index(0.5)
            y_pred_median = y_pred[:, median_idx]

            # Convert to numpy and transform back
            y_pred_np = y_pred_median.cpu().numpy()
            y_batch_np = y_batch.cpu().numpy()

            # Store predictions and targets
            all_preds.extend(y_pred_np)
            all_targets.extend(y_batch_np)

    # Convert to numpy arrays
    all_preds = np.array(all_preds).reshape(-1, 1)
    all_targets = np.array(all_targets).reshape(-1, 1)

    # Inverse transform to original scale
    all_preds = scaler_y.inverse_transform(all_preds)
    all_targets = scaler_y.inverse_transform(all_targets)

    # Calculate metrics
    metrics = calculate_metrics(all_targets, all_preds)

    return all_preds, all_targets, metrics

# Plot training and validation loss
def plot_losses(train_losses, val_losses):
    plt.figure(figsize=(10, 6))
    plt.plot(train_losses, label='Train Loss')
    plt.plot(val_losses, label='Validation Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.title('Training and Validation Loss')
    plt.legend()
    plt.grid(True)
    plt.savefig('lstm_loss_plot.png')
    plt.show()

# Plot predictions vs actual
def plot_predictions(y_true, y_pred, title='Predictions vs Actual'):
    plt.figure(figsize=(15, 8))
    plt.plot(y_true, label='Actual', color='blue')
    plt.plot(y_pred, label='Predicted', color='red', alpha=0.7)
    plt.xlabel('Time Step')
    plt.ylabel('Electricity Load (MWh)')
    plt.title(title)
    plt.legend()
    plt.grid(True)
    plt.savefig('lstm_predictions_plot.png')
    plt.show()

# Optuna objective function for hyperparameter tuning
def objective(trial):
    # Define the hyperparameters to optimize
    params = HyperParams()
    params.batch_size = trial.suggest_categorical('batch_size', [32, 64, 128])
    params.hidden_size = trial.suggest_categorical('hidden_size', [64, 128, 256])
    params.num_layers = trial.suggest_int('num_layers', 1, 3)
    params.dropout = trial.suggest_float('dropout', 0.1, 0.5)
    params.learning_rate = trial.suggest_loguniform('learning_rate', 1e-4, 1e-2)
    params.weight_decay = trial.suggest_loguniform('weight_decay', 1e-6, 1e-4)

    # Create data loaders
    train_loader, val_loader, _ = create_dataloaders(data_dict, params)

    # Create model
    input_dim = data_dict['train_features'].shape[1]
    output_dim = 1  # Predicting a single value

    model = LSTM(input_dim, output_dim, params).to(device)

    # Define optimizer and scheduler
    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=params.learning_rate,
        weight_decay=params.weight_decay
    )

    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode='min',
        factor=0.5,
        patience=5,
        verbose=True
    )

    # Train for fewer epochs during optimization
    num_epochs = 30
    patience = 7

    # Train model
    _, train_losses, val_losses = train(
        model,
        train_loader,
        val_loader,
        optimizer,
        scheduler,
        num_epochs,
        patience,
        device
    )

    # Return the best validation loss
    return min(val_losses)

# Main function
def main():
    # Load and preprocess data
    print("Loading and preprocessing data...")

    # Sample code to load data (replace with your actual data loading)
    df = pd.read_csv('/content/drive/MyDrive/thesis Final/Implementation/Panama Demand.csv')

    # Preprocess data
    global data_dict
    data_dict = preprocess_data(df)

    # Run Optuna optimization
    print("Running hyperparameter optimization with Optuna...")
    study = optuna.create_study(direction='minimize')
    study.optimize(objective, n_trials=20)

    # Get the best hyperparameters
    best_params = study.best_params
    print(f"Best hyperparameters: {best_params}")

    # Apply best hyperparameters
    params = HyperParams()
    params.batch_size = best_params['batch_size']
    params.hidden_size = best_params['hidden_size']
    params.num_layers = best_params['num_layers']
    params.dropout = best_params['dropout']
    params.learning_rate = best_params['learning_rate']
    params.weight_decay = best_params['weight_decay']

    # Create data loaders with best parameters
    train_loader, val_loader, test_loader = create_dataloaders(data_dict, params)

    # Create model with best parameters
    input_dim = data_dict['train_features'].shape[1]
    output_dim = 1  # Predicting a single value

    model = LSTM(input_dim, output_dim, params).to(device)

    # Define optimizer and scheduler
    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=params.learning_rate,
        weight_decay=params.weight_decay
    )

    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode='min',
        factor=0.5,
        patience=10,
        verbose=True
    )

    # Train the final model
    print("Training final model with best hyperparameters...")
    num_epochs = 100  # More epochs for final training
    patience = params.patience  # Early stopping patience

    final_model, train_losses, val_losses = train(
        model,
        train_loader,
        val_loader,
        optimizer,
        scheduler,
        num_epochs,
        patience,
        device
    )

    # Plot training and validation losses
    plot_losses(train_losses, val_losses)

    # Evaluate on test set
    print("Evaluating model on test set...")
    predictions, targets, metrics = evaluate(final_model, test_loader, data_dict['scaler_y'], device)

    # Print metrics
    print("\nTest Set Metrics:")
    for metric_name, metric_value in metrics.items():
        print(f"{metric_name}: {metric_value:.4f}")

    # Plot predictions vs actual
    print("Plotting predictions vs actual...")
    plot_predictions(targets.flatten(), predictions.flatten(), title='LSTM: Predictions vs Actual')

    # Plot a section of the predictions for better visibility
    # Take a smaller window (e.g., 2 weeks) for more detailed visualization
    window_size = 24 * 14  # 14 days assuming hourly data
    if len(targets) > window_size:
        start_idx = len(targets) // 2 - window_size // 2  # Center the window
        end_idx = start_idx + window_size

        # Plot the zoomed section
        plt.figure(figsize=(15, 8))
        plt.plot(targets[start_idx:end_idx], label='Actual', color='blue')
        plt.plot(predictions[start_idx:end_idx], label='Predicted', color='red', alpha=0.7)
        plt.xlabel('Time Step')
        plt.ylabel('Electricity Load (MWh)')
        plt.title('LSTM: Predictions vs Actual (2-Week Window)')
        plt.legend()
        plt.grid(True)
        plt.savefig('lstm_zoomed_predictions_plot.png')
        plt.show()

    # Save the model
    torch.save(final_model.state_dict(), 'lstm_model.pth')
    print("Model saved successfully!")

    return final_model, metrics


if __name__ == "__main__":
    # Replace with your actual data loading code
    df = pd.read_csv('/content/drive/MyDrive/thesis Final/Implementation/Panama Demand.csv')

    # Process the data
    data_dict = preprocess_data(df)

    # Run the main function
    model, metrics = main()

    print("\nFinal Test Set Metrics:")
    for metric_name, metric_value in metrics.items():
        print(f"{metric_name}: {metric_value:.4f}")
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error
import warnings
import optuna
from optuna.pruners import MedianPruner
from optuna.samplers import TPESampler
warnings.filterwarnings('ignore')

# Set seeds for reproducibility
SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

# Device configuration
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# Quantile levels for prediction
QUANTILES = [0.1, 0.5, 0.9]

# Hyperparameters class
class HyperParams:
    def __init__(self):
        # Default hyperparameters (will be optimized)
        self.batch_size = 32
        self.sequence_length = 24  # As specified in the requirements
        self.hidden_size = 128
        self.num_layers = 1
        self.dropout = 0.22269555755570092
        self.learning_rate = 0.002631739133963147
        self.weight_decay = 1.569536049001986e-05
        self.patience = 10  # Early stopping patience

# Custom dataset class
class ElectricityLoadDataset(Dataset):
    def __init__(self, features, targets, sequence_length):
        self.features = features
        self.targets = targets
        self.sequence_length = sequence_length

    def __len__(self):
        return len(self.features) - self.sequence_length

    def __getitem__(self, idx):
        x = self.features[idx:idx + self.sequence_length]
        y = self.targets[idx + self.sequence_length]
        return torch.FloatTensor(x), torch.FloatTensor([y])

# LSTM-only Model
class LSTM(nn.Module):
    def __init__(self, input_dim, output_dim, params):
        super(LSTM, self).__init__()

        self.params = params

        # LSTM layer
        self.lstm = nn.LSTM(
            input_size=input_dim,
            hidden_size=params.hidden_size,
            num_layers=params.num_layers,
            batch_first=True,
            dropout=params.dropout if params.num_layers > 1 else 0,
            bidirectional=True
        )

        # Dropout layer
        self.dropout = nn.Dropout(params.dropout)

        # Output layers for each quantile
        self.output_layers = nn.ModuleList([
            nn.Linear(params.hidden_size * 2, output_dim)  # * 2 for bidirectional
            for _ in QUANTILES
        ])

    def forward(self, x):
        # Apply LSTM
        x_lstm, (h_n, c_n) = self.lstm(x)

        # Get the output from the last time step
        x_last = x_lstm[:, -1, :]

        # Apply dropout
        x_last = self.dropout(x_last)

        # Apply output layers for each quantile
        quantile_outputs = [output_layer(x_last) for output_layer in self.output_layers]

        # Stack quantile outputs
        return torch.stack(quantile_outputs, dim=1)

# Pinball Loss for Quantile Regression
def pinball_loss(y_pred, y_true, quantiles=QUANTILES):
    losses = []
    for i, q in enumerate(quantiles):
        errors = y_true - y_pred[:, i]
        losses.append(torch.max(q * errors, (q - 1) * errors))

    return torch.mean(torch.stack(losses))

# Function to preprocess the Albania dataset
def preprocess_albania_data(df):
    # Convert Time to datetime
    df['datetime'] = pd.to_datetime(df['Time'])
    df.set_index('datetime', inplace=True)

    # Extract year for train/test split
    df['year'] = df.index.year

    # Handle missing values (if any)
    df.fillna(method='ffill', inplace=True)
    df.fillna(method='bfill', inplace=True)  # In case there are missing values at the beginning

    # Extract features and target
    target_column = 'Load'
    # Exclude 'Time', 'Time_double', 'year', and target from features
    feature_columns = [col for col in df.columns if col != target_column
                       and col != 'Time' and col != 'Time_double'
                       and col != 'year']

    # Create time features
    df['hour'] = df.index.hour
    df['month'] = df.index.month
    df['day'] = df.index.day
    df['dayofweek'] = df.index.dayofweek

    # Add sine and cosine transformations for cyclical features
    for col in ['hour', 'month', 'day', 'dayofweek']:
        if col == 'hour':
            max_val = 24
        elif col == 'month':
            max_val = 12
        elif col == 'dayofweek':
            max_val = 7
        else:  # day
            max_val = 31

        df[f'{col}_sin'] = np.sin(2 * np.pi * df[col] / max_val)
        df[f'{col}_cos'] = np.cos(2 * np.pi * df[col] / max_val)

        # Remove original columns if they're in feature_columns
        if col in feature_columns:
            feature_columns.remove(col)

        # Add cyclical features to feature columns
        feature_columns.extend([f'{col}_sin', f'{col}_cos'])

    # Create lags for the target variable
    for lag in [1, 2, 3, 6, 12, 24, 48, 72, 168]:  # 1h, 2h, 3h, 6h, 12h, 1d, 2d, 3d, 1w
        df[f'load_lag_{lag}'] = df[target_column].shift(lag)
        feature_columns.append(f'load_lag_{lag}')

    # Create rolling statistics
    for window in [6, 12, 24, 48, 72, 168]:  # 6h, 12h, 1d, 2d, 3d, 1w
        df[f'load_rolling_mean_{window}'] = df[target_column].rolling(window=window).mean().shift(1)
        df[f'load_rolling_std_{window}'] = df[target_column].rolling(window=window).std().shift(1)
        df[f'load_rolling_max_{window}'] = df[target_column].rolling(window=window).max().shift(1)
        df[f'load_rolling_min_{window}'] = df[target_column].rolling(window=window).min().shift(1)

        feature_columns.extend([
            f'load_rolling_mean_{window}',
            f'load_rolling_std_{window}',
            f'load_rolling_max_{window}',
            f'load_rolling_min_{window}'
        ])

    # Drop rows with NaN values (due to lagging and rolling operations)
    df.dropna(inplace=True)

    # Split data based on year: 2016-2018 as train, rest as test
    train_years = [2016, 2017, 2018]
    train_data = df[df['year'].isin(train_years)]
    test_data = df[~df['year'].isin(train_years)]

    # Split train data into train and validation (80/20 split)
    train_indices = train_data.index
    train_val_split = int(0.8 * len(train_indices))

    # Sort the indices to ensure chronological order
    train_indices = sorted(train_indices)

    train_data_final = train_data.loc[train_indices[:train_val_split]]
    val_data = train_data.loc[train_indices[train_val_split:]]

    # Remove 'year' column
    train_data_final = train_data_final.drop('year', axis=1)
    val_data = val_data.drop('year', axis=1)
    test_data = test_data.drop('year', axis=1)

    # Scale features
    scaler_X = StandardScaler()
    scaler_y = StandardScaler()

    # Fit on training data
    train_features = scaler_X.fit_transform(train_data_final[feature_columns])
    val_features = scaler_X.transform(val_data[feature_columns])
    test_features = scaler_X.transform(test_data[feature_columns])

    # Scale target separately
    train_target = scaler_y.fit_transform(train_data_final[[target_column]])
    val_target = scaler_y.transform(val_data[[target_column]])
    test_target = scaler_y.transform(test_data[[target_column]])

    # Flatten target arrays
    train_target = train_target.flatten()
    val_target = val_target.flatten()
    test_target = test_target.flatten()

    print(f"Train data shape: {train_features.shape}, {train_target.shape}")
    print(f"Validation data shape: {val_features.shape}, {val_target.shape}")
    print(f"Test data shape: {test_features.shape}, {test_target.shape}")

    # Create datasets
    return {
        'train_features': train_features,
        'train_target': train_target,
        'val_features': val_features,
        'val_target': val_target,
        'test_features': test_features,
        'test_target': test_target,
        'feature_columns': feature_columns,
        'target_column': target_column,
        'scaler_X': scaler_X,
        'scaler_y': scaler_y,
        'train_data': train_data_final,
        'val_data': val_data,
        'test_data': test_data
    }

# Create data loaders
def create_dataloaders(data_dict, params):
    # Create datasets
    train_dataset = ElectricityLoadDataset(
        data_dict['train_features'],
        data_dict['train_target'],
        params.sequence_length
    )

    val_dataset = ElectricityLoadDataset(
        data_dict['val_features'],
        data_dict['val_target'],
        params.sequence_length
    )

    test_dataset = ElectricityLoadDataset(
        data_dict['test_features'],
        data_dict['test_target'],
        params.sequence_length
    )

    # Create data loaders
    train_loader = DataLoader(
        train_dataset,
        batch_size=params.batch_size,
        shuffle=True,
        num_workers=4,
        pin_memory=True
    )

    val_loader = DataLoader(
        val_dataset,
        batch_size=params.batch_size,
        shuffle=False,
        num_workers=4,
        pin_memory=True
    )

    test_loader = DataLoader(
        test_dataset,
        batch_size=params.batch_size,
        shuffle=False,
        num_workers=4,
        pin_memory=True
    )

    return train_loader, val_loader, test_loader

# Define the evaluation metrics
def calculate_metrics(y_true, y_pred):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)

    # MSE
    mse = mean_squared_error(y_true, y_pred)
    # RMSE
    rmse = np.sqrt(mse)
    # MAE
    mae = mean_absolute_error(y_true, y_pred)
    # NRMSE (Normalized RMSE)
    nrmse = rmse / (np.max(y_true) - np.min(y_true))
    # MAPE (Mean Absolute Percentage Error)
    # Add small epsilon to avoid division by zero
    epsilon = 1e-8
    mape = np.mean(np.abs((y_true - y_pred) / (np.abs(y_true) + epsilon))) * 100

    return {
        'MSE': mse,
        'RMSE': rmse,
        'MAE': mae,
        'NRMSE': nrmse,
        'MAPE': mape
    }

# Training function
def train(model, train_loader, val_loader, optimizer, scheduler, num_epochs, patience, device, trial=None):
    train_losses = []
    val_losses = []
    best_val_loss = float('inf')
    patience_counter = 0
    best_model = None

    for epoch in range(num_epochs):
        # Training
        model.train()
        train_loss = 0

        for X_batch, y_batch in train_loader:
            X_batch, y_batch = X_batch.to(device), y_batch.to(device)

            # Forward pass
            y_pred = model(X_batch)

            # Calculate loss
            loss = pinball_loss(y_pred, y_batch)

            # Backward and optimize
            optimizer.zero_grad()
            loss.backward()

            # Apply gradient clipping to prevent exploding gradients
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)

            optimizer.step()

            train_loss += loss.item()

        train_loss /= len(train_loader)
        train_losses.append(train_loss)

        # Validation
        model.eval()
        val_loss = 0

        with torch.no_grad():
            for X_batch, y_batch in val_loader:
                X_batch, y_batch = X_batch.to(device), y_batch.to(device)

                # Forward pass
                y_pred = model(X_batch)

                # Calculate loss
                loss = pinball_loss(y_pred, y_batch)
                val_loss += loss.item()

        val_loss /= len(val_loader)
        val_losses.append(val_loss)

        # Update learning rate
        scheduler.step(val_loss)

        # Print progress (less verbose during hyperparameter search)
        if trial is None or epoch % 5 == 0:
            print(f'Epoch [{epoch+1}/{num_epochs}], Train Loss: {train_loss:.4f}, Val Loss: {val_loss:.4f}')

        # Report to Optuna for pruning if in a trial
        if trial is not None:
            trial.report(val_loss, epoch)

            # Handle pruning based on the intermediate value
            if trial.should_prune():
                raise optuna.exceptions.TrialPruned()

        # Check for early stopping
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            patience_counter = 0
            # Save best model
            best_model = model.state_dict().copy()
        else:
            patience_counter += 1
            if patience_counter >= patience:
                print(f'Early stopping at epoch {epoch+1}')
                break

    # Load best model
    model.load_state_dict(best_model)

    return model, train_losses, val_losses, best_val_loss

# Evaluation function
def evaluate(model, test_loader, scaler_y, device):
    model.eval()

    all_preds = []
    all_targets = []

    with torch.no_grad():
        for X_batch, y_batch in test_loader:
            X_batch, y_batch = X_batch.to(device), y_batch.to(device)

            # Forward pass
            y_pred = model(X_batch)

            # Extract median prediction (quantile 0.5)
            median_idx = QUANTILES.index(0.5)
            y_pred_median = y_pred[:, median_idx]

            # Convert to numpy and transform back
            y_pred_np = y_pred_median.cpu().numpy()
            y_batch_np = y_batch.cpu().numpy()

            # Store predictions and targets
            all_preds.extend(y_pred_np)
            all_targets.extend(y_batch_np)

    # Convert to numpy arrays
    all_preds = np.array(all_preds).reshape(-1, 1)
    all_targets = np.array(all_targets).reshape(-1, 1)

    # Inverse transform to original scale
    all_preds = scaler_y.inverse_transform(all_preds)
    all_targets = scaler_y.inverse_transform(all_targets)

    # Calculate metrics
    metrics = calculate_metrics(all_targets, all_preds)

    return all_preds, all_targets, metrics

# Plot training and validation loss
def plot_losses(train_losses, val_losses):
    plt.figure(figsize=(10, 6))
    plt.plot(train_losses, label='Train Loss')
    plt.plot(val_losses, label='Validation Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.title('Training and Validation Loss')
    plt.legend()
    plt.grid(True)
    plt.savefig('albania_lstm_loss_plot.png')
    plt.show()

# Plot predictions vs actual
def plot_predictions(y_true, y_pred, title='Predictions vs Actual'):
    plt.figure(figsize=(15, 8))
    plt.plot(y_true, label='Actual', color='blue')
    plt.plot(y_pred, label='Predicted', color='red', alpha=0.7)
    plt.xlabel('Time Step')
    plt.ylabel('Electricity Load (MWh)')
    plt.title(title)
    plt.legend()
    plt.grid(True)
    plt.savefig('albania_lstm_predictions_plot.png')
    plt.show()

# Define an objective function for Optuna
def objective(trial, data_dict, device, max_epochs=30):
    # Sample hyperparameters
    params = HyperParams()

    # Sample hyperparameters using Optuna
    params.batch_size = trial.suggest_categorical("batch_size", [16, 32, 64, 128, 256])
    params.hidden_size = trial.suggest_int("hidden_size", 32, 256, log=True)
    params.num_layers = trial.suggest_int("num_layers", 1, 3)
    params.dropout = trial.suggest_float("dropout", 0.0, 0.5)
    params.learning_rate = trial.suggest_float("learning_rate", 1e-5, 1e-2, log=True)
    params.weight_decay = trial.suggest_float("weight_decay", 1e-8, 1e-4, log=True)
    params.patience = trial.suggest_int("patience", 5, 15)

    # Create data loaders with the sampled batch size
    train_loader, val_loader, _ = create_dataloaders(data_dict, params)

    # Create the model with the sampled hyperparameters
    input_dim = data_dict['train_features'].shape[1]
    output_dim = 1  # Predicting a single value

    try:
        model = LSTM(input_dim, output_dim, params).to(device)

        # Define optimizer and scheduler
        optimizer = torch.optim.Adam(
            model.parameters(),
            lr=params.learning_rate,
            weight_decay=params.weight_decay
        )

        scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
            optimizer,
            mode='min',
            factor=0.5,
            patience=5,
            verbose=False  # Less verbose during hyperparameter search
        )

        # Train the model for a limited number of epochs during hyperparameter search
        _, _, _, best_val_loss = train(
            model,
            train_loader,
            val_loader,
            optimizer,
            scheduler,
            max_epochs,
            params.patience,
            device,
            trial=trial
        )

        return best_val_loss

    except Exception as e:
        print(f"Trial failed with error: {e}")
        # Return a high loss value to indicate failure
        return float('inf')

# Main function with hyperparameter optimization
def main():
    # Load and preprocess data
    print("Loading and preprocessing data...")
    df = pd.read_csv('/content/drive/MyDrive/thesis Final/Implementation/Albania_Dataset.csv')
    data_dict = preprocess_albania_data(df)

    # Run Optuna hyperparameter optimization
    print("Starting hyperparameter optimization with Optuna...")

    # Create Optuna study with pruning
    study = optuna.create_study(
        direction="minimize",
        sampler=TPESampler(seed=SEED),
        pruner=MedianPruner(n_startup_trials=5, n_warmup_steps=10)
    )

    # Optimize with 30 trials
    n_trials = 30
    study.optimize(
        lambda trial: objective(trial, data_dict, device),
        n_trials=n_trials
    )

    # Print optimization results
    print("Number of finished trials:", len(study.trials))
    print("Best trial:")
    trial = study.best_trial
    print("  Value (Validation Loss):", trial.value)
    print("  Params:")
    for key, value in trial.params.items():
        print(f"    {key}: {value}")

    # Use the best hyperparameters to train the final model
    print("\nTraining final model with best hyperparameters...")
    best_params = HyperParams()

    # Set the best hyperparameters
    for param_name, param_value in trial.params.items():
        setattr(best_params, param_name, param_value)

    # Create data loaders with the best hyperparameters
    train_loader, val_loader, test_loader = create_dataloaders(data_dict, best_params)

    # Create the model with the best hyperparameters
    input_dim = data_dict['train_features'].shape[1]
    output_dim = 1

    final_model = LSTM(input_dim, output_dim, best_params).to(device)

    # Define optimizer and scheduler
    optimizer = torch.optim.Adam(
        final_model.parameters(),
        lr=best_params.learning_rate,
        weight_decay=best_params.weight_decay
    )

    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode='min',
        factor=0.5,
        patience=5,
        verbose=True
    )

    # Train the final model for more epochs
    num_epochs = 100
    final_model, train_losses, val_losses, _ = train(
        final_model,
        train_loader,
        val_loader,
        optimizer,
        scheduler,
        num_epochs,
        best_params.patience,
        device
    )

    # Plot training and validation losses
    plot_losses(train_losses, val_losses)

    # Evaluate on test set
    print("Evaluating model on test set...")
    predictions, targets, metrics = evaluate(final_model, test_loader, data_dict['scaler_y'], device)

    # Print metrics
    print("\nTest Set Metrics:")
    for metric_name, metric_value in metrics.items():
        print(f"{metric_name}: {metric_value:.4f}")

    # Plot predictions vs actual
    print("Plotting predictions vs actual...")
    plot_predictions(targets.flatten(), predictions.flatten(), title='Albania LSTM: Predictions vs Actual')

    # Plot a section of the predictions for better visibility
    window_size = 24 * 14  # 14 days assuming hourly data
    if len(targets) > window_size:
        start_idx = len(targets) // 2 - window_size // 2  # Center the window
        end_idx = start_idx + window_size

        # Plot the zoomed section
        plt.figure(figsize=(15, 8))
        plt.plot(targets[start_idx:end_idx], label='Actual', color='blue')
        plt.plot(predictions[start_idx:end_idx], label='Predicted', color='red', alpha=0.7)
        plt.xlabel('Time Step')
        plt.ylabel('Electricity Load (MWh)')
        plt.title('Albania LSTM: Predictions vs Actual (2-Week Window)')
        plt.legend()
        plt.grid(True)
        plt.savefig('albania_lstm_zoomed_predictions_plot.png')
        plt.show()

    # Save the best hyperparameters
    import json
    best_params_dict = {attr: getattr(best_params, attr) for attr in dir(best_params)
                     if not callable(getattr(best_params, attr)) and not attr.startswith("__")}

    with open('best_hyperparameters.json', 'w') as f:
        json.dump(best_params_dict, f, indent=4)
    print("Best hyperparameters saved to 'best_hyperparameters.json'")

    # Save the model
    torch.save(final_model.state_dict(), 'albania_lstm_model.pth')
    print("Model saved successfully!")

    # Visualize hyperparameter importance and relationships if available
    try:
        # Parameter importance
        param_importances = optuna.visualization.plot_param_importances(study)
        param_importances.write_image('param_importances.png')
        print("Parameter importance plot saved to 'param_importances.png'")

        # Optimization history
        optimization_history = optuna.visualization.plot_optimization_history(study)
        optimization_history.write_image('optimization_history.png')
        print("Optimization history plot saved to 'optimization_history.png'")

        # Parameter relationships
        slice_plot = optuna.visualization.plot_slice(study)
        slice_plot.write_image('slice_plot.png')
        print("Parameter slice plot saved to 'slice_plot.png'")

        # Parallel coordinate plot
        parallel_plot = optuna.visualization.plot_parallel_coordinate(study)
        parallel_plot.write_image('parallel_coordinate.png')
        print("Parallel coordinate plot saved to 'parallel_coordinate.png'")
    except Exception as e:
        print(f"Visualization could not be created: {e}")

    return final_model, metrics, best_params

if __name__ == "__main__":
    # Try to use seaborn for better visualizations if available
    try:
        import seaborn as sns
        sns.set_theme(style="whitegrid")
    except ImportError:
        print("Seaborn not installed. Using matplotlib for visualizations.")

    # Run the main function
    model, metrics, best_params = main()

    print("\nFinal Test Set Metrics:")
    for metric_name, metric_value in metrics.items():
        print(f"{metric_name}: {metric_value:.4f}")

    print("\nBest Hyperparameters:")
    for param_name, param_value in vars(best_params).items():
        if not param_name.startswith('__'):
            print(f"{param_name}: {param_value}")
```
