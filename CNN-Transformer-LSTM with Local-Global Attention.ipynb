```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
import optuna
from optuna.trial import TrialState
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error
import warnings
import math
warnings.filterwarnings('ignore')

# Set seeds for reproducibility
SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

# Device configuration
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# Quantile levels for prediction
QUANTILES = [0.1, 0.5, 0.9]

# Hyperparameters (will be tuned by Optuna)
class HyperParams:
    def __init__(self):
        # Default values, will be overridden by Optuna
        self.batch_size = 64
        self.sequence_length = 24  # As specified in the requirements
        self.hidden_size = 128
        self.num_layers = 2
        self.dropout = 0.2
        self.learning_rate = 0.001
        self.weight_decay = 1e-5
        self.cnn_filters = 64
        self.cnn_kernel_size = 3
        self.nhead = 8  # Number of heads in transformer
        self.transformer_dim = 128
        self.transformer_layers = 2
        self.patience = 10  # Early stopping patience
        self.local_window_size = 6  # Size of local attention window (in timesteps)

# Custom dataset class
class ElectricityLoadDataset(Dataset):
    def __init__(self, features, targets, sequence_length):
        self.features = features
        self.targets = targets
        self.sequence_length = sequence_length

    def __len__(self):
        return len(self.features) - self.sequence_length

    def __getitem__(self, idx):
        x = self.features[idx:idx + self.sequence_length]
        y = self.targets[idx + self.sequence_length]
        return torch.FloatTensor(x), torch.FloatTensor([y])

# CNN Block
class CNNBlock(nn.Module):
    def __init__(self, in_channels, out_channels, kernel_size, dropout):
        super(CNNBlock, self).__init__()
        self.conv1 = nn.Conv1d(in_channels, out_channels, kernel_size, padding=kernel_size//2)
        self.bn1 = nn.BatchNorm1d(out_channels)
        self.conv2 = nn.Conv1d(out_channels, out_channels, kernel_size, padding=kernel_size//2)
        self.bn2 = nn.BatchNorm1d(out_channels)
        self.relu = nn.ReLU()
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # x shape: [batch, sequence, features]
        x = x.permute(0, 2, 1)  # Change to [batch, features, sequence]
        x = self.relu(self.bn1(self.conv1(x)))
        x = self.dropout(x)
        x = self.relu(self.bn2(self.conv2(x)))
        x = self.dropout(x)
        x = x.permute(0, 2, 1)  # Change back to [batch, sequence, features]
        return x

# Positional Encoding for Transformer
class PositionalEncoding(nn.Module):
    def __init__(self, d_model, dropout=0.1, max_len=5000):
        super(PositionalEncoding, self).__init__()
        self.dropout = nn.Dropout(p=dropout)

        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-np.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        pe = pe.unsqueeze(0)
        self.register_buffer('pe', pe)

    def forward(self, x):
        x = x + self.pe[:, :x.size(1), :]
        return self.dropout(x)

# Transformer Block
class TransformerBlock(nn.Module):
    def __init__(self, d_model, nhead, dim_feedforward, dropout, num_layers):
        super(TransformerBlock, self).__init__()
        encoder_layers = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=nhead,
            dim_feedforward=dim_feedforward,
            dropout=dropout,
            batch_first=True  # For [batch, seq, feature] input
        )
        self.transformer_encoder = nn.TransformerEncoder(encoder_layers, num_layers=num_layers)
        self.pos_encoder = PositionalEncoding(d_model, dropout)

    def forward(self, x):
        # Add positional encoding
        x = self.pos_encoder(x)
        # Apply transformer
        return self.transformer_encoder(x)

# Local-Global Attention Module
class LocalGlobalAttention(nn.Module):
    def __init__(self, embed_dim, num_heads, local_window_size, dropout=0.1):
        super(LocalGlobalAttention, self).__init__()
        self.embed_dim = embed_dim
        self.num_heads = num_heads
        self.head_dim = embed_dim // num_heads
        self.local_window_size = local_window_size

        assert self.head_dim * num_heads == embed_dim, "embed_dim must be divisible by num_heads"

        # Local attention (focused on nearby timesteps)
        self.local_attention = nn.MultiheadAttention(
            embed_dim=embed_dim,
            num_heads=num_heads,
            dropout=dropout,
            batch_first=True
        )

        # Global attention (considering the entire sequence)
        self.global_attention = nn.MultiheadAttention(
            embed_dim=embed_dim,
            num_heads=num_heads,
            dropout=dropout,
            batch_first=True
        )

        # Gating mechanism to balance local and global information
        self.gate = nn.Sequential(
            nn.Linear(embed_dim * 2, embed_dim),
            nn.Sigmoid()
        )

        # Layer normalization for stability
        self.norm1 = nn.LayerNorm(embed_dim)
        self.norm2 = nn.LayerNorm(embed_dim)
        self.norm_final = nn.LayerNorm(embed_dim)

        # Output projection
        self.output_projection = nn.Linear(embed_dim, embed_dim)
        self.dropout = nn.Dropout(dropout)

    def _generate_local_mask(self, seq_len, local_window_size, device):
        """Generate a mask for local attention with specified window size"""
        # Initialize mask with zeros (masked positions)
        mask = torch.zeros(seq_len, seq_len, device=device)

        # Fill diagonal band with ones (visible positions)
        for i in range(seq_len):
            # Define window boundaries
            window_start = max(0, i - local_window_size // 2)
            window_end = min(seq_len, i + local_window_size // 2 + 1)
            # Set visible positions to 1
            mask[i, window_start:window_end] = 1

        return mask

    def forward(self, x):
        # x shape: [batch_size, sequence_length, embed_dim]
        batch_size, seq_len, _ = x.shape

        # Apply normalization
        normed_x = self.norm1(x)

        # Generate local attention mask
        local_mask = self._generate_local_mask(seq_len, self.local_window_size, x.device)

        # Apply local attention (with window mask)
        local_output, _ = self.local_attention(
            query=normed_x,
            key=normed_x,
            value=normed_x,
            attn_mask=local_mask,
            need_weights=False
        )

        # Apply global attention (full sequence)
        global_output, _ = self.global_attention(
            query=self.norm2(x),
            key=self.norm2(x),
            value=self.norm2(x),
            need_weights=False
        )

        # Calculate adaptive gate values to balance local and global information
        combined = torch.cat([local_output, global_output], dim=-1)
        gate_values = self.gate(combined)

        # Mix local and global features based on gate values
        mixed_output = gate_values * local_output + (1 - gate_values) * global_output

        # Apply output projection and residual connection
        output = x + self.dropout(self.output_projection(mixed_output))

        # Final normalization
        output = self.norm_final(output)

        return output

# CNN-Transformer-LSTM Model with Local-Global Attention
class CNNTransformerLSTM(nn.Module):
    def __init__(self, input_dim, output_dim, params):
        super(CNNTransformerLSTM, self).__init__()

        self.params = params

        # CNN to extract local patterns
        self.cnn = CNNBlock(
            in_channels=input_dim,
            out_channels=params.cnn_filters,
            kernel_size=params.cnn_kernel_size,
            dropout=params.dropout
        )

        # Linear layer to map CNN output to transformer dimension
        self.cnn_to_transformer = nn.Linear(params.cnn_filters, params.transformer_dim)

        # Transformer for capturing global dependencies
        self.transformer = TransformerBlock(
            d_model=params.transformer_dim,
            nhead=params.nhead,
            dim_feedforward=params.transformer_dim * 4,
            dropout=params.dropout,
            num_layers=params.transformer_layers
        )

        # LSTM for temporal dependencies
        self.lstm = nn.LSTM(
            input_size=params.transformer_dim,
            hidden_size=params.hidden_size,
            num_layers=params.num_layers,
            batch_first=True,
            dropout=params.dropout if params.num_layers > 1 else 0,
            bidirectional=True
        )

        # Replace standard attention with Local-Global Attention
        self.local_global_attention = LocalGlobalAttention(
            embed_dim=params.hidden_size * 2,  # * 2 for bidirectional LSTM
            num_heads=params.nhead,
            local_window_size=params.local_window_size,
            dropout=params.dropout
        )

        # Dropout layer
        self.dropout = nn.Dropout(params.dropout)

        # Output layers for each quantile
        self.output_layers = nn.ModuleList([
            nn.Linear(params.hidden_size * 2, output_dim)  # * 2 for bidirectional
            for _ in QUANTILES
        ])

    def forward(self, x):
        # Apply CNN
        x_cnn = self.cnn(x)

        # Map to transformer dimension
        x_transformer_input = self.cnn_to_transformer(x_cnn)

        # Apply Transformer
        x_transformer = self.transformer(x_transformer_input)

        # Apply LSTM
        x_lstm, (h_n, c_n) = self.lstm(x_transformer)

        # Apply Local-Global attention to LSTM outputs
        x_attn = self.local_global_attention(x_lstm)

        # Get the last time step for prediction
        x_last = x_attn[:, -1, :]

        # Apply dropout
        x_last = self.dropout(x_last)

        # Apply output layers for each quantile
        quantile_outputs = [output_layer(x_last) for output_layer in self.output_layers]

        # Stack quantile outputs
        return torch.stack(quantile_outputs, dim=1)

# Pinball Loss for Quantile Regression
def pinball_loss(y_pred, y_true, quantiles=QUANTILES):
    losses = []
    for i, q in enumerate(quantiles):
        errors = y_true - y_pred[:, i]
        losses.append(torch.max(q * errors, (q - 1) * errors))

    return torch.mean(torch.stack(losses))

# Function to preprocess data
def preprocess_data(df):
    # Convert datetime to index
    df['datetime'] = pd.to_datetime(df['datetime'])
    df.set_index('datetime', inplace=True)

    # Extract year for train/val/test split
    df['year'] = df.index.year

    # Handle missing values (if any)
    df.fillna(method='ffill', inplace=True)
    df.fillna(method='bfill', inplace=True)  # In case there are missing values at the beginning

    # Extract features and target
    target_column = 'nat_demand'
    feature_columns = [col for col in df.columns if col != target_column and col != 'year']

    # Create time features
    df['hour'] = df.index.hour
    df['month'] = df.index.month
    df['day'] = df.index.day

    # Add sine and cosine transformations for cyclical features
    for col in ['hour', 'month', 'day']:
        if col == 'hour':
            max_val = 24
        elif col == 'month':
            max_val = 12
        else:  # day
            max_val = 31

        df[f'{col}_sin'] = np.sin(2 * np.pi * df[col] / max_val)
        df[f'{col}_cos'] = np.cos(2 * np.pi * df[col] / max_val)

        # Remove original columns
        feature_columns.remove(col) if col in feature_columns else None

    # Add cyclical features to feature columns
    for col in ['hour', 'month', 'day']:
        feature_columns.extend([f'{col}_sin', f'{col}_cos'])

    # Create lags for the target variable
    for lag in [1, 2, 3, 6, 12, 24, 48, 72, 168]:  # 1h, 2h, 3h, 6h, 12h, 1d, 2d, 3d, 1w
        df[f'nat_demand_lag_{lag}'] = df[target_column].shift(lag)
        feature_columns.append(f'nat_demand_lag_{lag}')

    # Create rolling statistics
    for window in [6, 12, 24, 48, 72, 168]:  # 6h, 12h, 1d, 2d, 3d, 1w
        df[f'nat_demand_rolling_mean_{window}'] = df[target_column].rolling(window=window).mean().shift(1)
        df[f'nat_demand_rolling_std_{window}'] = df[target_column].rolling(window=window).std().shift(1)
        df[f'nat_demand_rolling_max_{window}'] = df[target_column].rolling(window=window).max().shift(1)
        df[f'nat_demand_rolling_min_{window}'] = df[target_column].rolling(window=window).min().shift(1)

        feature_columns.extend([
            f'nat_demand_rolling_mean_{window}',
            f'nat_demand_rolling_std_{window}',
            f'nat_demand_rolling_max_{window}',
            f'nat_demand_rolling_min_{window}'
        ])

    # Drop rows with NaN values (due to lagging and rolling operations)
    df.dropna(inplace=True)

    # Split data based on year
    train_data = df[df['year'].isin([2015, 2016, 2017])]
    val_data = df[df['year'] == 2018]
    test_data = df[~df['year'].isin([2015, 2016, 2017, 2018])]

    # Remove 'year' column
    train_data = train_data.drop('year', axis=1)
    val_data = val_data.drop('year', axis=1)
    test_data = test_data.drop('year', axis=1)

    # Scale features
    scaler_X = StandardScaler()
    scaler_y = StandardScaler()

    # Fit on training data
    train_features = scaler_X.fit_transform(train_data[feature_columns])
    val_features = scaler_X.transform(val_data[feature_columns])
    test_features = scaler_X.transform(test_data[feature_columns])

    # Scale target separately
    train_target = scaler_y.fit_transform(train_data[[target_column]])
    val_target = scaler_y.transform(val_data[[target_column]])
    test_target = scaler_y.transform(test_data[[target_column]])

    # Flatten target arrays
    train_target = train_target.flatten()
    val_target = val_target.flatten()
    test_target = test_target.flatten()

    # Create datasets
    return {
        'train_features': train_features,
        'train_target': train_target,
        'val_features': val_features,
        'val_target': val_target,
        'test_features': test_features,
        'test_target': test_target,
        'feature_columns': feature_columns,
        'target_column': target_column,
        'scaler_X': scaler_X,
        'scaler_y': scaler_y,
        'train_data': train_data,
        'val_data': val_data,
        'test_data': test_data
    }

# Create data loaders
def create_dataloaders(data_dict, params):
    # Create datasets
    train_dataset = ElectricityLoadDataset(
        data_dict['train_features'],
        data_dict['train_target'],
        params.sequence_length
    )

    val_dataset = ElectricityLoadDataset(
        data_dict['val_features'],
        data_dict['val_target'],
        params.sequence_length
    )

    test_dataset = ElectricityLoadDataset(
        data_dict['test_features'],
        data_dict['test_target'],
        params.sequence_length
    )

    # Create data loaders
    train_loader = DataLoader(
        train_dataset,
        batch_size=params.batch_size,
        shuffle=True,
        num_workers=4,
        pin_memory=True
    )

    val_loader = DataLoader(
        val_dataset,
        batch_size=params.batch_size,
        shuffle=False,
        num_workers=4,
        pin_memory=True
    )

    test_loader = DataLoader(
        test_dataset,
        batch_size=params.batch_size,
        shuffle=False,
        num_workers=4,
        pin_memory=True
    )

    return train_loader, val_loader, test_loader

# Define the evaluation metrics
def calculate_metrics(y_true, y_pred):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)

    # MSE
    mse = mean_squared_error(y_true, y_pred)
    # RMSE
    rmse = np.sqrt(mse)
    # MAE
    mae = mean_absolute_error(y_true, y_pred)
    # NRMSE (Normalized RMSE)
    nrmse = rmse / (np.max(y_true) - np.min(y_true))
    # MAPE (Mean Absolute Percentage Error)
    mape = np.mean(np.abs((y_true - y_pred) / y_true)) * 100

    return {
        'MSE': mse,
        'RMSE': rmse,
        'MAE': mae,
        'NRMSE': nrmse,
        'MAPE': mape
    }

# Training function
def train(model, train_loader, val_loader, optimizer, scheduler, num_epochs, patience, device):
    train_losses = []
    val_losses = []
    best_val_loss = float('inf')
    patience_counter = 0
    best_model = None

    for epoch in range(num_epochs):
        # Training
        model.train()
        train_loss = 0

        for X_batch, y_batch in train_loader:
            X_batch, y_batch = X_batch.to(device), y_batch.to(device)

            # Forward pass
            y_pred = model(X_batch)

            # Calculate loss
            loss = pinball_loss(y_pred, y_batch)

            # Backward and optimize
            optimizer.zero_grad()
            loss.backward()

            # Apply gradient clipping to prevent exploding gradients
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)

            optimizer.step()

            train_loss += loss.item()

        train_loss /= len(train_loader)
        train_losses.append(train_loss)

        # Validation
        model.eval()
        val_loss = 0

        with torch.no_grad():
            for X_batch, y_batch in val_loader:
                X_batch, y_batch = X_batch.to(device), y_batch.to(device)

                # Forward pass
                y_pred = model(X_batch)

                # Calculate loss
                loss = pinball_loss(y_pred, y_batch)
                val_loss += loss.item()

        val_loss /= len(val_loader)
        val_losses.append(val_loss)

        # Update learning rate
        scheduler.step(val_loss)

        # Print progress
        print(f'Epoch [{epoch+1}/{num_epochs}], Train Loss: {train_loss:.4f}, Val Loss: {val_loss:.4f}')

        # Check for early stopping
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            patience_counter = 0
            # Save best model
            best_model = model.state_dict().copy()
        else:
            patience_counter += 1
            if patience_counter >= patience:
                print(f'Early stopping at epoch {epoch+1}')
                break

    # Load best model
    model.load_state_dict(best_model)

    return model, train_losses, val_losses

# Evaluation function
def evaluate(model, test_loader, scaler_y, device):
    model.eval()

    all_preds = []
    all_targets = []

    with torch.no_grad():
        for X_batch, y_batch in test_loader:
            X_batch, y_batch = X_batch.to(device), y_batch.to(device)

            # Forward pass
            y_pred = model(X_batch)

            # Extract median prediction (quantile 0.5)
            median_idx = QUANTILES.index(0.5)
            y_pred_median = y_pred[:, median_idx]

            # Convert to numpy and transform back
            y_pred_np = y_pred_median.cpu().numpy()
            y_batch_np = y_batch.cpu().numpy()

            # Store predictions and targets
            all_preds.extend(y_pred_np)
            all_targets.extend(y_batch_np)

    # Convert to numpy arrays
    all_preds = np.array(all_preds).reshape(-1, 1)
    all_targets = np.array(all_targets).reshape(-1, 1)

    # Inverse transform to original scale
    all_preds = scaler_y.inverse_transform(all_preds)
    all_targets = scaler_y.inverse_transform(all_targets)

    # Calculate metrics
    metrics = calculate_metrics(all_targets, all_preds)

    return all_preds, all_targets, metrics

# Plot training and validation loss
def plot_losses(train_losses, val_losses):
    plt.figure(figsize=(10, 6))
    plt.plot(train_losses, label='Train Loss')
    plt.plot(val_losses, label='Validation Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.title('Training and Validation Loss')
    plt.legend()
    plt.grid(True)
    plt.savefig('local_global_attention_loss_plot.png')
    plt.show()

# Plot predictions vs actual
def plot_predictions(y_true, y_pred, title='Predictions vs Actual'):
    plt.figure(figsize=(15, 8))
    plt.plot(y_true, label='Actual', color='blue')
    plt.plot(y_pred, label='Predicted', color='red', alpha=0.7)
    plt.xlabel('Time Step')
    plt.ylabel('Electricity Load (MWh)')
    plt.title(title)
    plt.legend()
    plt.grid(True)
    plt.savefig('local_global_attention_predictions_plot.png')
    plt.show()

# Optuna objective function for hyperparameter tuning
def objective(trial):
    # Define the hyperparameters to optimize
    params = HyperParams()
    params.batch_size = trial.suggest_categorical('batch_size', [32, 64, 128])
    params.hidden_size = trial.suggest_categorical('hidden_size', [64, 128, 256])
    params.num_layers = trial.suggest_int('num_layers', 1, 3)
    params.dropout = trial.suggest_float('dropout', 0.1, 0.5)
    params.learning_rate = trial.suggest_loguniform('learning_rate', 1e-4, 1e-2)
    params.weight_decay = trial.suggest_loguniform('weight_decay', 1e-6, 1e-4)
    params.cnn_filters = trial.suggest_categorical('cnn_filters', [32, 64, 128])
    params.cnn_kernel_size = trial.suggest_categorical('cnn_kernel_size', [3, 5, 7])
    params.nhead = trial.suggest_categorical('nhead', [4, 8])
    params.transformer_dim = trial.suggest_categorical('transformer_dim', [64, 128, 256])
    params.transformer_layers = trial.suggest_int('transformer_layers', 1, 3)

    # Add local window size for Local-Global Attention
    params.local_window_size = trial.suggest_int('local_window_size', 4, 12)

    # Create data loaders
    train_loader, val_loader, _ = create_dataloaders(data_dict, params)

    # Create model
    input_dim = data_dict['train_features'].shape[1]
    output_dim = 1  # Predicting a single value

    model = CNNTransformerLSTM(input_dim, output_dim, params).to(device)

    # Define optimizer and scheduler
    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=params.learning_rate,
        weight_decay=params.weight_decay
    )

    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode='min',
        factor=0.5,
        patience=5,
        verbose=True
    )

    # Train for fewer epochs during optimization
    num_epochs = 30
    patience = 7

    # Train model
    _, train_losses, val_losses = train(
        model,
        train_loader,
        val_loader,
        optimizer,
        scheduler,
        num_epochs,
        patience,
        device
    )

    # Return the best validation loss
    return min(val_losses)

# Main function
def main():
    # Load and preprocess data
    print("Loading and preprocessing data...")

    # Sample code to load data (replace with your actual data loading)
    df = pd.read_csv('/content/drive/MyDrive/thesis Final/Implementation/Panama Demand.csv')

    # Preprocess data
    global data_dict
    data_dict = preprocess_data(df)

    # Run Optuna optimization
    print("Running hyperparameter optimization with Optuna...")
    study = optuna.create_study(direction='minimize')
    study.optimize(objective, n_trials=20)

    # Get the best hyperparameters
    best_params = study.best_params
    print(f"Best hyperparameters: {best_params}")

    # Apply best hyperparameters
    params = HyperParams()
    params.batch_size = best_params['batch_size']
    params.hidden_size = best_params['hidden_size']
    params.num_layers = best_params['num_layers']
    params.dropout = best_params['dropout']
    params.learning_rate = best_params['learning_rate']
    params.weight_decay = best_params['weight_decay']
    params.cnn_filters = best_params['cnn_filters']
    params.cnn_kernel_size = best_params['cnn_kernel_size']
    params.nhead = best_params['nhead']
    params.transformer_dim = best_params['transformer_dim']
    params.transformer_layers = best_params['transformer_layers']
    params.local_window_size = best_params['local_window_size']

    # Create data loaders with best parameters
    train_loader, val_loader, test_loader = create_dataloaders(data_dict, params)

    # Create model with best parameters
    input_dim = data_dict['train_features'].shape[1]
    output_dim = 1  # Predicting a single value

    model = CNNTransformerLSTM(input_dim, output_dim, params).to(device)

    # Define optimizer and scheduler
    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=params.learning_rate,
        weight_decay=params.weight_decay
    )

    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode='min',
        factor=0.5,
        patience=10,
        verbose=True
    )

    # Train the final model
    print("Training final model with best hyperparameters...")
    num_epochs = 100  # More epochs for final training
    patience = params.patience  # Early stopping patience

    final_model, train_losses, val_losses = train(
        model,
        train_loader,
        val_loader,
        optimizer,
        scheduler,
        num_epochs,
        patience,
        device
    )

    # Plot training and validation losses
    plot_losses(train_losses, val_losses)

    # Evaluate on test set
    print("Evaluating model on test set...")
    predictions, targets, metrics = evaluate(final_model, test_loader, data_dict['scaler_y'], device)

    # Print metrics
    print("\nTest Set Metrics:")
    for metric_name, metric_value in metrics.items():
        print(f"{metric_name}: {metric_value:.4f}")

    # Plot predictions vs actual
    print("Plotting predictions vs actual...")
    plot_predictions(targets.flatten(), predictions.flatten(), title='CNN-Transformer-LSTM with Local-Global Attention: Predictions vs Actual')

    # Plot a section of the predictions for better visibility
    # Take a smaller window (e.g., 2 weeks) for more detailed visualization
    window_size = 24 * 14  # 14 days assuming hourly data
    if len(targets) > window_size:
        start_idx = len(targets) // 2 - window_size // 2  # Center the window
        end_idx = start_idx + window_size

        # Plot the zoomed section
        plt.figure(figsize=(15, 8))
        plt.plot(targets[start_idx:end_idx], label='Actual', color='blue')
        plt.plot(predictions[start_idx:end_idx], label='Predicted', color='red', alpha=0.7)
        plt.xlabel('Time Step')
        plt.ylabel('Electricity Load (MWh)')
        plt.title('CNN-Transformer-LSTM with Local-Global Attention: Predictions vs Actual (2-Week Window)')
        plt.legend()
        plt.grid(True)
        plt.savefig('local_global_attention_zoomed_predictions_plot.png')
        plt.show()

    # Save the model
    torch.save(final_model.state_dict(), 'local_global_attention_model.pth')
    print("Model saved successfully!")

    return final_model, metrics


if __name__ == "__main__":
    # Import seaborn for better visualizations
    try:
        import seaborn as sns
        sns.set_theme(style="whitegrid")
    except ImportError:
        print("Seaborn not installed. Using matplotlib for visualizations.")

    # Replace with your actual data loading code
    df = pd.read_csv('/content/drive/MyDrive/thesis Final/Implementation/Panama Demand.csv')

    # Process the data
    data_dict = preprocess_data(df)

    # Run the main function
    model, metrics = main()

    # Optionally visualize attention weights
    # visualize_attention_weights(model, create_dataloaders(data_dict, model.params)[2], device)

    print("\nFinal Test Set Metrics:")
    for metric_name, metric_value in metrics.items():
        print(f"{metric_name}: {metric_value:.4f}")
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error
import warnings
import math
warnings.filterwarnings('ignore')

# Set seeds for reproducibility
SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

# Device configuration
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# Quantile levels for prediction
QUANTILES = [0.1, 0.5, 0.9]

# Hyperparameters (best values from Optuna search)
class HyperParams:
    def __init__(self):
        # Best hyperparameters from Optuna
        self.batch_size = 32
        self.sequence_length = 24  # As specified in the requirements
        self.hidden_size = 128
        self.num_layers = 1
        self.dropout = 0.12923361001405317
        self.learning_rate = 0.000641285932448859
        self.weight_decay = 1.8179096220981544e-05
        self.cnn_filters = 32
        self.cnn_kernel_size = 5
        self.nhead = 8
        self.transformer_dim = 128
        self.transformer_layers = 3
        self.patience = 10  # Early stopping patience
        self.local_window_size = 8  # Size of local attention window (in timesteps)

# Custom dataset class
class ElectricityLoadDataset(Dataset):
    def __init__(self, features, targets, sequence_length):
        self.features = features
        self.targets = targets
        self.sequence_length = sequence_length

    def __len__(self):
        return len(self.features) - self.sequence_length

    def __getitem__(self, idx):
        x = self.features[idx:idx + self.sequence_length]
        y = self.targets[idx + self.sequence_length]
        return torch.FloatTensor(x), torch.FloatTensor([y])

# CNN Block
class CNNBlock(nn.Module):
    def __init__(self, in_channels, out_channels, kernel_size, dropout):
        super(CNNBlock, self).__init__()
        self.conv1 = nn.Conv1d(in_channels, out_channels, kernel_size, padding=kernel_size//2)
        self.bn1 = nn.BatchNorm1d(out_channels)
        self.conv2 = nn.Conv1d(out_channels, out_channels, kernel_size, padding=kernel_size//2)
        self.bn2 = nn.BatchNorm1d(out_channels)
        self.relu = nn.ReLU()
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # x shape: [batch, sequence, features]
        x = x.permute(0, 2, 1)  # Change to [batch, features, sequence]
        x = self.relu(self.bn1(self.conv1(x)))
        x = self.dropout(x)
        x = self.relu(self.bn2(self.conv2(x)))
        x = self.dropout(x)
        x = x.permute(0, 2, 1)  # Change back to [batch, sequence, features]
        return x

# Positional Encoding for Transformer
class PositionalEncoding(nn.Module):
    def __init__(self, d_model, dropout=0.1, max_len=5000):
        super(PositionalEncoding, self).__init__()
        self.dropout = nn.Dropout(p=dropout)

        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-np.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        pe = pe.unsqueeze(0)
        self.register_buffer('pe', pe)

    def forward(self, x):
        x = x + self.pe[:, :x.size(1), :]
        return self.dropout(x)

# Transformer Block
class TransformerBlock(nn.Module):
    def __init__(self, d_model, nhead, dim_feedforward, dropout, num_layers):
        super(TransformerBlock, self).__init__()
        encoder_layers = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=nhead,
            dim_feedforward=dim_feedforward,
            dropout=dropout,
            batch_first=True  # For [batch, seq, feature] input
        )
        self.transformer_encoder = nn.TransformerEncoder(encoder_layers, num_layers=num_layers)
        self.pos_encoder = PositionalEncoding(d_model, dropout)

    def forward(self, x):
        # Add positional encoding
        x = self.pos_encoder(x)
        # Apply transformer
        return self.transformer_encoder(x)

# Local-Global Attention Module
class LocalGlobalAttention(nn.Module):
    def __init__(self, embed_dim, num_heads, local_window_size, dropout=0.1):
        super(LocalGlobalAttention, self).__init__()
        self.embed_dim = embed_dim
        self.num_heads = num_heads
        self.head_dim = embed_dim // num_heads
        self.local_window_size = local_window_size

        assert self.head_dim * num_heads == embed_dim, "embed_dim must be divisible by num_heads"

        # Local attention (focused on nearby timesteps)
        self.local_attention = nn.MultiheadAttention(
            embed_dim=embed_dim,
            num_heads=num_heads,
            dropout=dropout,
            batch_first=True
        )

        # Global attention (considering the entire sequence)
        self.global_attention = nn.MultiheadAttention(
            embed_dim=embed_dim,
            num_heads=num_heads,
            dropout=dropout,
            batch_first=True
        )

        # Gating mechanism to balance local and global information
        self.gate = nn.Sequential(
            nn.Linear(embed_dim * 2, embed_dim),
            nn.Sigmoid()
        )

        # Layer normalization for stability
        self.norm1 = nn.LayerNorm(embed_dim)
        self.norm2 = nn.LayerNorm(embed_dim)
        self.norm_final = nn.LayerNorm(embed_dim)

        # Output projection
        self.output_projection = nn.Linear(embed_dim, embed_dim)
        self.dropout = nn.Dropout(dropout)

    def _generate_local_mask(self, seq_len, local_window_size, device):
        """Generate a mask for local attention with specified window size"""
        # Initialize mask with zeros (masked positions)
        mask = torch.zeros(seq_len, seq_len, device=device)

        # Fill diagonal band with ones (visible positions)
        for i in range(seq_len):
            # Define window boundaries
            window_start = max(0, i - local_window_size // 2)
            window_end = min(seq_len, i + local_window_size // 2 + 1)
            # Set visible positions to 1
            mask[i, window_start:window_end] = 1

        return mask

    def forward(self, x):
        # x shape: [batch_size, sequence_length, embed_dim]
        batch_size, seq_len, _ = x.shape

        # Apply normalization
        normed_x = self.norm1(x)

        # Generate local attention mask
        local_mask = self._generate_local_mask(seq_len, self.local_window_size, x.device)

        # Apply local attention (with window mask)
        local_output, _ = self.local_attention(
            query=normed_x,
            key=normed_x,
            value=normed_x,
            attn_mask=local_mask,
            need_weights=False
        )

        # Apply global attention (full sequence)
        global_output, _ = self.global_attention(
            query=self.norm2(x),
            key=self.norm2(x),
            value=self.norm2(x),
            need_weights=False
        )

        # Calculate adaptive gate values to balance local and global information
        combined = torch.cat([local_output, global_output], dim=-1)
        gate_values = self.gate(combined)

        # Mix local and global features based on gate values
        mixed_output = gate_values * local_output + (1 - gate_values) * global_output

        # Apply output projection and residual connection
        output = x + self.dropout(self.output_projection(mixed_output))

        # Final normalization
        output = self.norm_final(output)

        return output

# CNN-Transformer-LSTM Model with Local-Global Attention
class CNNTransformerLSTM(nn.Module):
    def __init__(self, input_dim, output_dim, params):
        super(CNNTransformerLSTM, self).__init__()

        self.params = params

        # CNN to extract local patterns
        self.cnn = CNNBlock(
            in_channels=input_dim,
            out_channels=params.cnn_filters,
            kernel_size=params.cnn_kernel_size,
            dropout=params.dropout
        )

        # Linear layer to map CNN output to transformer dimension
        self.cnn_to_transformer = nn.Linear(params.cnn_filters, params.transformer_dim)

        # Transformer for capturing global dependencies
        self.transformer = TransformerBlock(
            d_model=params.transformer_dim,
            nhead=params.nhead,
            dim_feedforward=params.transformer_dim * 4,
            dropout=params.dropout,
            num_layers=params.transformer_layers
        )

        # LSTM for temporal dependencies
        self.lstm = nn.LSTM(
            input_size=params.transformer_dim,
            hidden_size=params.hidden_size,
            num_layers=params.num_layers,
            batch_first=True,
            dropout=params.dropout if params.num_layers > 1 else 0,
            bidirectional=True
        )

        # Replace standard attention with Local-Global Attention
        self.local_global_attention = LocalGlobalAttention(
            embed_dim=params.hidden_size * 2,  # * 2 for bidirectional LSTM
            num_heads=params.nhead,
            local_window_size=params.local_window_size,
            dropout=params.dropout
        )

        # Dropout layer
        self.dropout = nn.Dropout(params.dropout)

        # Output layers for each quantile
        self.output_layers = nn.ModuleList([
            nn.Linear(params.hidden_size * 2, output_dim)  # * 2 for bidirectional
            for _ in QUANTILES
        ])

    def forward(self, x):
        # Apply CNN
        x_cnn = self.cnn(x)

        # Map to transformer dimension
        x_transformer_input = self.cnn_to_transformer(x_cnn)

        # Apply Transformer
        x_transformer = self.transformer(x_transformer_input)

        # Apply LSTM
        x_lstm, (h_n, c_n) = self.lstm(x_transformer)

        # Apply Local-Global attention to LSTM outputs
        x_attn = self.local_global_attention(x_lstm)

        # Get the last time step for prediction
        x_last = x_attn[:, -1, :]

        # Apply dropout
        x_last = self.dropout(x_last)

        # Apply output layers for each quantile
        quantile_outputs = [output_layer(x_last) for output_layer in self.output_layers]

        # Stack quantile outputs
        return torch.stack(quantile_outputs, dim=1)

# Pinball Loss for Quantile Regression
def pinball_loss(y_pred, y_true, quantiles=QUANTILES):
    losses = []
    for i, q in enumerate(quantiles):
        errors = y_true - y_pred[:, i]
        losses.append(torch.max(q * errors, (q - 1) * errors))

    return torch.mean(torch.stack(losses))

# Function to preprocess data
def preprocess_data(df):
    # Convert datetime to index
    df['datetime'] = pd.to_datetime(df['datetime'])
    df.set_index('datetime', inplace=True)

    # Extract year for train/val/test split
    df['year'] = df.index.year

    # Handle missing values (if any)
    df.fillna(method='ffill', inplace=True)
    df.fillna(method='bfill', inplace=True)  # In case there are missing values at the beginning

    # Extract features and target
    target_column = 'nat_demand'
    feature_columns = [col for col in df.columns if col != target_column and col != 'year']

    # Create time features
    df['hour'] = df.index.hour
    df['month'] = df.index.month
    df['day'] = df.index.day

    # Add sine and cosine transformations for cyclical features
    for col in ['hour', 'month', 'day']:
        if col == 'hour':
            max_val = 24
        elif col == 'month':
            max_val = 12
        else:  # day
            max_val = 31

        df[f'{col}_sin'] = np.sin(2 * np.pi * df[col] / max_val)
        df[f'{col}_cos'] = np.cos(2 * np.pi * df[col] / max_val)

        # Remove original columns
        feature_columns.remove(col) if col in feature_columns else None

    # Add cyclical features to feature columns
    for col in ['hour', 'month', 'day']:
        feature_columns.extend([f'{col}_sin', f'{col}_cos'])

    # Create lags for the target variable
    for lag in [1, 2, 3, 6, 12, 24, 48, 72, 168]:  # 1h, 2h, 3h, 6h, 12h, 1d, 2d, 3d, 1w
        df[f'nat_demand_lag_{lag}'] = df[target_column].shift(lag)
        feature_columns.append(f'nat_demand_lag_{lag}')

    # Create rolling statistics
    for window in [6, 12, 24, 48, 72, 168]:  # 6h, 12h, 1d, 2d, 3d, 1w
        df[f'nat_demand_rolling_mean_{window}'] = df[target_column].rolling(window=window).mean().shift(1)
        df[f'nat_demand_rolling_std_{window}'] = df[target_column].rolling(window=window).std().shift(1)
        df[f'nat_demand_rolling_max_{window}'] = df[target_column].rolling(window=window).max().shift(1)
        df[f'nat_demand_rolling_min_{window}'] = df[target_column].rolling(window=window).min().shift(1)

        feature_columns.extend([
            f'nat_demand_rolling_mean_{window}',
            f'nat_demand_rolling_std_{window}',
            f'nat_demand_rolling_max_{window}',
            f'nat_demand_rolling_min_{window}'
        ])

    # Drop rows with NaN values (due to lagging and rolling operations)
    df.dropna(inplace=True)

    # Split data based on year
    train_data = df[df['year'].isin([2015, 2016, 2017])]
    val_data = df[df['year'] == 2018]
    test_data = df[~df['year'].isin([2015, 2016, 2017, 2018])]

    # Remove 'year' column
    train_data = train_data.drop('year', axis=1)
    val_data = val_data.drop('year', axis=1)
    test_data = test_data.drop('year', axis=1)

    # Scale features
    scaler_X = StandardScaler()
    scaler_y = StandardScaler()

    # Fit on training data
    train_features = scaler_X.fit_transform(train_data[feature_columns])
    val_features = scaler_X.transform(val_data[feature_columns])
    test_features = scaler_X.transform(test_data[feature_columns])

    # Scale target separately
    train_target = scaler_y.fit_transform(train_data[[target_column]])
    val_target = scaler_y.transform(val_data[[target_column]])
    test_target = scaler_y.transform(test_data[[target_column]])

    # Flatten target arrays
    train_target = train_target.flatten()
    val_target = val_target.flatten()
    test_target = test_target.flatten()

    # Create datasets
    return {
        'train_features': train_features,
        'train_target': train_target,
        'val_features': val_features,
        'val_target': val_target,
        'test_features': test_features,
        'test_target': test_target,
        'feature_columns': feature_columns,
        'target_column': target_column,
        'scaler_X': scaler_X,
        'scaler_y': scaler_y,
        'train_data': train_data,
        'val_data': val_data,
        'test_data': test_data
    }

# Create data loaders
def create_dataloaders(data_dict, params):
    # Create datasets
    train_dataset = ElectricityLoadDataset(
        data_dict['train_features'],
        data_dict['train_target'],
        params.sequence_length
    )

    val_dataset = ElectricityLoadDataset(
        data_dict['val_features'],
        data_dict['val_target'],
        params.sequence_length
    )

    test_dataset = ElectricityLoadDataset(
        data_dict['test_features'],
        data_dict['test_target'],
        params.sequence_length
    )

    # Create data loaders
    train_loader = DataLoader(
        train_dataset,
        batch_size=params.batch_size,
        shuffle=True,
        num_workers=4,
        pin_memory=True
    )

    val_loader = DataLoader(
        val_dataset,
        batch_size=params.batch_size,
        shuffle=False,
        num_workers=4,
        pin_memory=True
    )

    test_loader = DataLoader(
        test_dataset,
        batch_size=params.batch_size,
        shuffle=False,
        num_workers=4,
        pin_memory=True
    )

    return train_loader, val_loader, test_loader

# Define the evaluation metrics
def calculate_metrics(y_true, y_pred):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)

    # MSE
    mse = mean_squared_error(y_true, y_pred)
    # RMSE
    rmse = np.sqrt(mse)
    # MAE
    mae = mean_absolute_error(y_true, y_pred)
    # NRMSE (Normalized RMSE)
    nrmse = rmse / (np.max(y_true) - np.min(y_true))
    # MAPE (Mean Absolute Percentage Error)
    mape = np.mean(np.abs((y_true - y_pred) / y_true)) * 100

    return {
        'MSE': mse,
        'RMSE': rmse,
        'MAE': mae,
        'NRMSE': nrmse,
        'MAPE': mape
    }

# Training function
def train(model, train_loader, val_loader, optimizer, scheduler, num_epochs, patience, device):
    train_losses = []
    val_losses = []
    best_val_loss = float('inf')
    patience_counter = 0
    best_model = None

    for epoch in range(num_epochs):
        # Training
        model.train()
        train_loss = 0

        for X_batch, y_batch in train_loader:
            X_batch, y_batch = X_batch.to(device), y_batch.to(device)

            # Forward pass
            y_pred = model(X_batch)

            # Calculate loss
            loss = pinball_loss(y_pred, y_batch)

            # Backward and optimize
            optimizer.zero_grad()
            loss.backward()

            # Apply gradient clipping to prevent exploding gradients
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)

            optimizer.step()

            train_loss += loss.item()

        train_loss /= len(train_loader)
        train_losses.append(train_loss)

        # Validation
        model.eval()
        val_loss = 0

        with torch.no_grad():
            for X_batch, y_batch in val_loader:
                X_batch, y_batch = X_batch.to(device), y_batch.to(device)

                # Forward pass
                y_pred = model(X_batch)

                # Calculate loss
                loss = pinball_loss(y_pred, y_batch)
                val_loss += loss.item()

        val_loss /= len(val_loader)
        val_losses.append(val_loss)

        # Update learning rate
        scheduler.step(val_loss)

        # Print progress
        print(f'Epoch [{epoch+1}/{num_epochs}], Train Loss: {train_loss:.4f}, Val Loss: {val_loss:.4f}')

        # Check for early stopping
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            patience_counter = 0
            # Save best model
            best_model = model.state_dict().copy()
        else:
            patience_counter += 1
            if patience_counter >= patience:
                print(f'Early stopping at epoch {epoch+1}')
                break

    # Load best model
    model.load_state_dict(best_model)

    return model, train_losses, val_losses

# Evaluation function with quantile predictions
def evaluate_quantiles(model, test_loader, scaler_y, device):
    model.eval()

    all_preds_q10 = []
    all_preds_q50 = []
    all_preds_q90 = []
    all_targets = []

    with torch.no_grad():
        for X_batch, y_batch in test_loader:
            X_batch, y_batch = X_batch.to(device), y_batch.to(device)

            # Forward pass
            y_pred = model(X_batch)  # Shape: [batch_size, num_quantiles, 1]

            # Extract predictions for each quantile
            y_pred_q10 = y_pred[:, 0].cpu().numpy()  # 10% quantile
            y_pred_q50 = y_pred[:, 1].cpu().numpy()  # 50% quantile (median)
            y_pred_q90 = y_pred[:, 2].cpu().numpy()  # 90% quantile

            y_batch_np = y_batch.cpu().numpy()

            # Store predictions and targets
            all_preds_q10.extend(y_pred_q10)
            all_preds_q50.extend(y_pred_q50)
            all_preds_q90.extend(y_pred_q90)
            all_targets.extend(y_batch_np)

    # Convert to numpy arrays
    all_preds_q10 = np.array(all_preds_q10).reshape(-1, 1)
    all_preds_q50 = np.array(all_preds_q50).reshape(-1, 1)
    all_preds_q90 = np.array(all_preds_q90).reshape(-1, 1)
    all_targets = np.array(all_targets).reshape(-1, 1)

    # Inverse transform to original scale
    all_preds_q10 = scaler_y.inverse_transform(all_preds_q10)
    all_preds_q50 = scaler_y.inverse_transform(all_preds_q50)
    all_preds_q90 = scaler_y.inverse_transform(all_preds_q90)
    all_targets = scaler_y.inverse_transform(all_targets)

    # Calculate metrics for the median prediction (q50)
    metrics = calculate_metrics(all_targets, all_preds_q50)

    return {
        'q10': all_preds_q10.flatten(),
        'q50': all_preds_q50.flatten(),
        'q90': all_preds_q90.flatten(),
        'targets': all_targets.flatten(),
        'metrics': metrics
    }

# Plot training and validation loss
def plot_losses(train_losses, val_losses):
    plt.figure(figsize=(10, 6))
    plt.plot(train_losses, label='Train Loss')
    plt.plot(val_losses, label='Validation Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.title('Training and Validation Loss')
    plt.legend()
    plt.grid(True)
    plt.savefig('local_global_attention_loss_plot.png')
    plt.show()

# Plot predictions vs actual with quantile range
def plot_predictions_with_quantiles(results, title='Predictions vs Actual with Quantile Range'):
    plt.figure(figsize=(15, 8))

    # Plot actual values
    plt.plot(results['targets'], label='Actual', color='blue', linewidth=2)

    # Plot median prediction (q50)
    plt.plot(results['q50'], label='Median (q50) Prediction', color='red', alpha=0.7, linewidth=2)

    # Plot the quantile range (q10-q90) representing 80% prediction interval
    plt.fill_between(
        range(len(results['q10'])),
        results['q10'],
        results['q90'],
        alpha=0.2,
        color='gray',
        label='80% Confidence Interval (q10-q90)'
    )

    plt.xlabel('Time Step')
    plt.ylabel('Electricity Load (MWh)')
    plt.title(title)
    plt.legend()
    plt.grid(True)
    plt.savefig('predictions_with_quantile_range.png')
    plt.show()

# Plot a zoomed section with quantile range
def plot_zoomed_predictions_with_quantiles(results, window_size=24*14, title='Predictions vs Actual with Quantile Range (2-Week Window)'):
    # Take a smaller window for more detailed visualization
    if len(results['targets']) > window_size:
        start_idx = len(results['targets']) // 2 - window_size // 2  # Center the window
        end_idx = start_idx + window_size

        plt.figure(figsize=(15, 8))

        # Plot actual values
        plt.plot(
            results['targets'][start_idx:end_idx],
            label='Actual',
            color='blue',
            linewidth=2
        )

        # Plot median prediction (q50)
        plt.plot(
            results['q50'][start_idx:end_idx],
            label='Median (q50) Prediction',
            color='red',
            alpha=0.7,
            linewidth=2
        )

        # Plot the quantile range (q10-q90)
        plt.fill_between(
            range(window_size),
            results['q10'][start_idx:end_idx],
            results['q90'][start_idx:end_idx],
            alpha=0.2,
            color='red',
            label='80% Confidence Interval (q10-q90)'
        )

        plt.xlabel('Time Step')
        plt.ylabel('Electricity Load (MWh)')
        plt.title(title)
        plt.legend()
        plt.grid(True)
        plt.savefig('zoomed_predictions_with_quantile_range.png')
        plt.show()

# Main function
def main():
    # Load and preprocess data
    print("Loading and preprocessing data...")

    # Sample code to load data (replace with your actual data loading)
    df = pd.read_csv('/content/drive/MyDrive/thesis Final/Implementation/Panama Demand.csv')

    # Preprocess data
    data_dict = preprocess_data(df)

    # Use best hyperparameters
    params = HyperParams()

    # Create data loaders with best parameters
    train_loader, val_loader, test_loader = create_dataloaders(data_dict, params)

    # Create model with best parameters
    input_dim = data_dict['train_features'].shape[1]
    output_dim = 1  # Predicting a single value

    model = CNNTransformerLSTM(input_dim, output_dim, params).to(device)

    # Define optimizer and scheduler
    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=params.learning_rate,
        weight_decay=params.weight_decay
    )

    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode='min',
        factor=0.5,
        patience=10,
        verbose=True
    )

    # Train the final model
    print("Training final model with best hyperparameters...")
    num_epochs = 100  # More epochs for final training
    patience = params.patience  # Early stopping patience

    final_model, train_losses, val_losses = train(
        model,
        train_loader,
        val_loader,
        optimizer,
        scheduler,
        num_epochs,
        patience,
        device
    )

    # Plot training and validation losses
    plot_losses(train_losses, val_losses)

    # Evaluate on test set with quantile predictions
    print("Evaluating model on test set...")
    results = evaluate_quantiles(final_model, test_loader, data_dict['scaler_y'], device)

    # Print metrics
    print("\nTest Set Metrics:")
    for metric_name, metric_value in results['metrics'].items():
        print(f"{metric_name}: {metric_value:.4f}")

    # Plot predictions vs actual with quantile range
    print("Plotting predictions vs actual with quantile ranges...")
    plot_predictions_with_quantiles(
        results,
        title='CNN-Transformer-LSTM with Local-Global Attention: Predictions with 80% Confidence Interval'
    )

    # Plot a zoomed section
    print("Plotting zoomed section with quantile ranges...")
    plot_zoomed_predictions_with_quantiles(
        results,
        window_size=24 * 14,  # 14 days of hourly data
        title='CNN-Transformer-LSTM: Predictions with 80% Confidence Interval (2-Week Window)'
    )

    # Calculate coverage of the 80% prediction interval
    q10_array = results['q10']
    q90_array = results['q90']
    targets_array = results['targets']

    # Count how many actual values fall within the prediction interval
    within_interval = np.logical_and(
        targets_array >= q10_array,
        targets_array <= q90_array
    )
    coverage = np.mean(within_interval) * 100

    print(f"\nActual coverage of 80% prediction interval: {coverage:.2f}%")

    # Calculate average interval width
    interval_width = np.mean(q90_array - q10_array)
    print(f"Average width of 80% prediction interval: {interval_width:.2f} MWh")

    # Plot interval coverage over time
    plt.figure(figsize=(15, 6))
    window_size = 24*7  # 7-day rolling window
    within_interval_series = pd.Series(within_interval)
    rolling_coverage = within_interval_series.rolling(window_size).mean() * 100

    plt.plot(rolling_coverage, label=f'{window_size//24}-day Rolling Coverage')
    plt.axhline(y=80, color='r', linestyle='--', label='Target (80%)')
    plt.xlabel('Time Step')
    plt.ylabel('Coverage (%)')
    plt.title('Rolling Coverage of 80% Prediction Interval')
    plt.legend()
    plt.grid(True)
    plt.savefig('interval_coverage_over_time.png')
    plt.show()

    # Save the model
    torch.save(final_model.state_dict(), 'local_global_attention_model.pth')
    print("Model saved successfully!")

    return final_model, results


if __name__ == "__main__":
    # Import seaborn for better visualizations
    try:
        import seaborn as sns
        sns.set_theme(style="whitegrid")
    except ImportError:
        print("Seaborn not installed. Using matplotlib for visualizations.")

    # Load data
    df = pd.read_csv('/content/drive/MyDrive/thesis Final/Implementation/Panama Demand.csv')

    # Process the data and run main
    model, results = main()

    # Additional analysis of the results

    # 1. Visualize the distribution of prediction errors for the median forecast
    errors = results['targets'] - results['q50']

    plt.figure(figsize=(12, 6))
    plt.subplot(1, 2, 1)
    sns.histplot(errors, kde=True)
    plt.title('Distribution of Prediction Errors')
    plt.axvline(x=0, color='r', linestyle='--')

    plt.subplot(1, 2, 2)
    sns.boxplot(y=errors)
    plt.title('Boxplot of Prediction Errors')
    plt.tight_layout()
    plt.savefig('error_distribution.png')
    plt.show()

    # 2. Calculate and plot pinball losses for each quantile
    def calculate_pinball_loss(y_true, y_pred, quantile):
        errors = y_true - y_pred
        return np.mean(np.maximum(quantile * errors, (quantile - 1) * errors))

    pinball_q10 = calculate_pinball_loss(results['targets'], results['q10'], 0.1)
    pinball_q50 = calculate_pinball_loss(results['targets'], results['q50'], 0.5)
    pinball_q90 = calculate_pinball_loss(results['targets'], results['q90'], 0.9)

    print(f"\nPinball Loss by Quantile:")
    print(f"Q10 (10% quantile): {pinball_q10:.4f}")
    print(f"Q50 (50% quantile): {pinball_q50:.4f}")
    print(f"Q90 (90% quantile): {pinball_q90:.4f}")

    # Plot pinball losses
    plt.figure(figsize=(8, 6))
    plt.bar(['Q10 (10%)', 'Q50 (50%)', 'Q90 (90%)'],
            [pinball_q10, pinball_q50, pinball_q90])
    plt.title('Pinball Loss by Quantile')
    plt.ylabel('Loss Value')
    plt.grid(axis='y')
    plt.savefig('pinball_loss_by_quantile.png')
    plt.show()

    # 3. Visualize uncertainty across different time periods
    # Calculate prediction interval width (uncertainty)
    interval_width = results['q90'] - results['q10']

    # Assuming test_data from data_dict has the same length as results
    # We can use interval_width and associate with datetime
    if 'test_data' in locals():
        test_dates = test_data.index[-len(interval_width):]
        uncertainty_df = pd.DataFrame({
            'date': test_dates,
            'interval_width': interval_width,
            'hour': test_dates.hour,
            'day_of_week': test_dates.dayofweek,
            'month': test_dates.month
        })

        # Plot uncertainty by hour of day
        plt.figure(figsize=(15, 10))

        plt.subplot(2, 2, 1)
        sns.boxplot(x='hour', y='interval_width', data=uncertainty_df)
        plt.title('Prediction Uncertainty by Hour of Day')
        plt.xlabel('Hour of Day')
        plt.ylabel('Interval Width (MWh)')

        plt.subplot(2, 2, 2)
        sns.boxplot(x='day_of_week', y='interval_width', data=uncertainty_df)
        plt.title('Prediction Uncertainty by Day of Week')
        plt.xlabel('Day of Week (0=Monday, 6=Sunday)')
        plt.ylabel('Interval Width (MWh)')

        plt.subplot(2, 2, 3)
        sns.boxplot(x='month', y='interval_width', data=uncertainty_df)
        plt.title('Prediction Uncertainty by Month')
        plt.xlabel('Month')
        plt.ylabel('Interval Width (MWh)')

        plt.tight_layout()
        plt.savefig('uncertainty_by_time_period.png')
        plt.show()
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
import optuna
from optuna.trial import TrialState
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error
import warnings
import math
warnings.filterwarnings('ignore')

# Set seeds for reproducibility
SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

# Device configuration
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# Quantile levels for prediction
QUANTILES = [0.1, 0.5, 0.9]

# Hyperparameters (will be tuned by Optuna)
class HyperParams:
    def __init__(self):
        # Default values, will be overridden by Optuna
        self.batch_size = 64
        self.sequence_length = 24  # As specified in the requirements
        self.hidden_size = 128
        self.num_layers = 2
        self.dropout = 0.2
        self.learning_rate = 0.001
        self.weight_decay = 1e-5
        self.cnn_filters = 64
        self.cnn_kernel_size = 3
        self.nhead = 8  # Number of heads in transformer
        self.transformer_dim = 128
        self.transformer_layers = 2
        self.patience = 10  # Early stopping patience
        self.local_window_size = 6  # Size of local attention window (in timesteps)

# Custom dataset class
class ElectricityLoadDataset(Dataset):
    def __init__(self, features, targets, sequence_length):
        self.features = features
        self.targets = targets
        self.sequence_length = sequence_length

    def __len__(self):
        return len(self.features) - self.sequence_length

    def __getitem__(self, idx):
        x = self.features[idx:idx + self.sequence_length]
        y = self.targets[idx + self.sequence_length]
        return torch.FloatTensor(x), torch.FloatTensor([y])

# CNN Block
class CNNBlock(nn.Module):
    def __init__(self, in_channels, out_channels, kernel_size, dropout):
        super(CNNBlock, self).__init__()
        self.conv1 = nn.Conv1d(in_channels, out_channels, kernel_size, padding=kernel_size//2)
        self.bn1 = nn.BatchNorm1d(out_channels)
        self.conv2 = nn.Conv1d(out_channels, out_channels, kernel_size, padding=kernel_size//2)
        self.bn2 = nn.BatchNorm1d(out_channels)
        self.relu = nn.ReLU()
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # x shape: [batch, sequence, features]
        x = x.permute(0, 2, 1)  # Change to [batch, features, sequence]
        x = self.relu(self.bn1(self.conv1(x)))
        x = self.dropout(x)
        x = self.relu(self.bn2(self.conv2(x)))
        x = self.dropout(x)
        x = x.permute(0, 2, 1)  # Change back to [batch, sequence, features]
        return x

# Positional Encoding for Transformer
class PositionalEncoding(nn.Module):
    def __init__(self, d_model, dropout=0.1, max_len=5000):
        super(PositionalEncoding, self).__init__()
        self.dropout = nn.Dropout(p=dropout)

        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-np.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        pe = pe.unsqueeze(0)
        self.register_buffer('pe', pe)

    def forward(self, x):
        x = x + self.pe[:, :x.size(1), :]
        return self.dropout(x)

# Transformer Block
class TransformerBlock(nn.Module):
    def __init__(self, d_model, nhead, dim_feedforward, dropout, num_layers):
        super(TransformerBlock, self).__init__()
        encoder_layers = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=nhead,
            dim_feedforward=dim_feedforward,
            dropout=dropout,
            batch_first=True  # For [batch, seq, feature] input
        )
        self.transformer_encoder = nn.TransformerEncoder(encoder_layers, num_layers=num_layers)
        self.pos_encoder = PositionalEncoding(d_model, dropout)

    def forward(self, x):
        # Add positional encoding
        x = self.pos_encoder(x)
        # Apply transformer
        return self.transformer_encoder(x)

# Local-Global Attention Module
class LocalGlobalAttention(nn.Module):
    def __init__(self, embed_dim, num_heads, local_window_size, dropout=0.1):
        super(LocalGlobalAttention, self).__init__()
        self.embed_dim = embed_dim
        self.num_heads = num_heads
        self.head_dim = embed_dim // num_heads
        self.local_window_size = local_window_size

        assert self.head_dim * num_heads == embed_dim, "embed_dim must be divisible by num_heads"

        # Local attention (focused on nearby timesteps)
        self.local_attention = nn.MultiheadAttention(
            embed_dim=embed_dim,
            num_heads=num_heads,
            dropout=dropout,
            batch_first=True
        )

        # Global attention (considering the entire sequence)
        self.global_attention = nn.MultiheadAttention(
            embed_dim=embed_dim,
            num_heads=num_heads,
            dropout=dropout,
            batch_first=True
        )

        # Gating mechanism to balance local and global information
        self.gate = nn.Sequential(
            nn.Linear(embed_dim * 2, embed_dim),
            nn.Sigmoid()
        )

        # Layer normalization for stability
        self.norm1 = nn.LayerNorm(embed_dim)
        self.norm2 = nn.LayerNorm(embed_dim)
        self.norm_final = nn.LayerNorm(embed_dim)

        # Output projection
        self.output_projection = nn.Linear(embed_dim, embed_dim)
        self.dropout = nn.Dropout(dropout)

    def _generate_local_mask(self, seq_len, local_window_size, device):
        """Generate a mask for local attention with specified window size"""
        # Initialize mask with zeros (masked positions)
        mask = torch.zeros(seq_len, seq_len, device=device)

        # Fill diagonal band with ones (visible positions)
        for i in range(seq_len):
            # Define window boundaries
            window_start = max(0, i - local_window_size // 2)
            window_end = min(seq_len, i + local_window_size // 2 + 1)
            # Set visible positions to 1
            mask[i, window_start:window_end] = 1

        return mask

    def forward(self, x):
        # x shape: [batch_size, sequence_length, embed_dim]
        batch_size, seq_len, _ = x.shape

        # Apply normalization
        normed_x = self.norm1(x)

        # Generate local attention mask
        local_mask = self._generate_local_mask(seq_len, self.local_window_size, x.device)

        # Apply local attention (with window mask)
        local_output, _ = self.local_attention(
            query=normed_x,
            key=normed_x,
            value=normed_x,
            attn_mask=local_mask,
            need_weights=False
        )

        # Apply global attention (full sequence)
        global_output, _ = self.global_attention(
            query=self.norm2(x),
            key=self.norm2(x),
            value=self.norm2(x),
            need_weights=False
        )

        # Calculate adaptive gate values to balance local and global information
        combined = torch.cat([local_output, global_output], dim=-1)
        gate_values = self.gate(combined)

        # Mix local and global features based on gate values
        mixed_output = gate_values * local_output + (1 - gate_values) * global_output

        # Apply output projection and residual connection
        output = x + self.dropout(self.output_projection(mixed_output))

        # Final normalization
        output = self.norm_final(output)

        return output

# CNN-Transformer-LSTM Model with Local-Global Attention
class CNNTransformerLSTM(nn.Module):
    def __init__(self, input_dim, output_dim, params):
        super(CNNTransformerLSTM, self).__init__()

        self.params = params

        # CNN to extract local patterns
        self.cnn = CNNBlock(
            in_channels=input_dim,
            out_channels=params.cnn_filters,
            kernel_size=params.cnn_kernel_size,
            dropout=params.dropout
        )

        # Linear layer to map CNN output to transformer dimension
        self.cnn_to_transformer = nn.Linear(params.cnn_filters, params.transformer_dim)

        # Transformer for capturing global dependencies
        self.transformer = TransformerBlock(
            d_model=params.transformer_dim,
            nhead=params.nhead,
            dim_feedforward=params.transformer_dim * 4,
            dropout=params.dropout,
            num_layers=params.transformer_layers
        )

        # LSTM for temporal dependencies
        self.lstm = nn.LSTM(
            input_size=params.transformer_dim,
            hidden_size=params.hidden_size,
            num_layers=params.num_layers,
            batch_first=True,
            dropout=params.dropout if params.num_layers > 1 else 0,
            bidirectional=True
        )

        # Replace standard attention with Local-Global Attention
        self.local_global_attention = LocalGlobalAttention(
            embed_dim=params.hidden_size * 2,  # * 2 for bidirectional LSTM
            num_heads=params.nhead,
            local_window_size=params.local_window_size,
            dropout=params.dropout
        )

        # Dropout layer
        self.dropout = nn.Dropout(params.dropout)

        # Output layers for each quantile
        self.output_layers = nn.ModuleList([
            nn.Linear(params.hidden_size * 2, output_dim)  # * 2 for bidirectional
            for _ in QUANTILES
        ])

    def forward(self, x):
        # Apply CNN
        x_cnn = self.cnn(x)

        # Map to transformer dimension
        x_transformer_input = self.cnn_to_transformer(x_cnn)

        # Apply Transformer
        x_transformer = self.transformer(x_transformer_input)

        # Apply LSTM
        x_lstm, (h_n, c_n) = self.lstm(x_transformer)

        # Apply Local-Global attention to LSTM outputs
        x_attn = self.local_global_attention(x_lstm)

        # Get the last time step for prediction
        x_last = x_attn[:, -1, :]

        # Apply dropout
        x_last = self.dropout(x_last)

        # Apply output layers for each quantile
        quantile_outputs = [output_layer(x_last) for output_layer in self.output_layers]

        # Stack quantile outputs
        return torch.stack(quantile_outputs, dim=1)

# Pinball Loss for Quantile Regression
def pinball_loss(y_pred, y_true, quantiles=QUANTILES):
    losses = []
    for i, q in enumerate(quantiles):
        errors = y_true - y_pred[:, i]
        losses.append(torch.max(q * errors, (q - 1) * errors))

    return torch.mean(torch.stack(losses))

# Function to preprocess data
def preprocess_data(df, test_size=0.15, val_size=0.15):
    # Extract target variable
    target_column = 'target_variable'
    feature_columns = [col for col in df.columns if col != target_column]

    # Create time features assuming data is sequential
    df['hour'] = np.arange(len(df)) % 24
    df['day'] = (np.arange(len(df)) // 24) % 7
    df['month'] = (np.arange(len(df)) // (24 * 30)) % 12 + 1  # Approximation

    # Add sine and cosine transformations for cyclical features
    for col in ['hour', 'day', 'month']:
        if col == 'hour':
            max_val = 24
        elif col == 'day':
            max_val = 7
        else:  # month
            max_val = 12

        df[f'{col}_sin'] = np.sin(2 * np.pi * df[col] / max_val)
        df[f'{col}_cos'] = np.cos(2 * np.pi * df[col] / max_val)

    # Add these new columns to feature columns and remove original time columns
    feature_columns.extend([f'{col}_sin' for col in ['hour', 'day', 'month']])
    feature_columns.extend([f'{col}_cos' for col in ['hour', 'day', 'month']])

    # Create lags for the target variable
    for lag in [1, 2, 3, 6, 12, 24, 48, 72, 168]:  # 1h, 2h, 3h, 6h, 12h, 1d, 2d, 3d, 1w
        df[f'target_lag_{lag}'] = df[target_column].shift(lag)
        feature_columns.append(f'target_lag_{lag}')

    # Create rolling statistics
    for window in [6, 12, 24, 48, 72, 168]:  # 6h, 12h, 1d, 2d, 3d, 1w
        df[f'target_rolling_mean_{window}'] = df[target_column].rolling(window=window).mean().shift(1)
        df[f'target_rolling_std_{window}'] = df[target_column].rolling(window=window).std().shift(1)
        df[f'target_rolling_max_{window}'] = df[target_column].rolling(window=window).max().shift(1)
        df[f'target_rolling_min_{window}'] = df[target_column].rolling(window=window).min().shift(1)

        feature_columns.extend([
            f'target_rolling_mean_{window}',
            f'target_rolling_std_{window}',
            f'target_rolling_max_{window}',
            f'target_rolling_min_{window}'
        ])

    # Drop rows with NaN values (due to lagging and rolling operations)
    df.dropna(inplace=True)

    # Split data based on chronological order (since no year column in the dataset)
    total_samples = len(df)
    test_samples = int(test_size * total_samples)
    val_samples = int(val_size * total_samples)
    train_samples = total_samples - test_samples - val_samples

    train_data = df.iloc[:train_samples]
    val_data = df.iloc[train_samples:train_samples+val_samples]
    test_data = df.iloc[train_samples+val_samples:]

    # Scale features
    scaler_X = StandardScaler()
    scaler_y = StandardScaler()

    # Fit on training data
    train_features = scaler_X.fit_transform(train_data[feature_columns])
    val_features = scaler_X.transform(val_data[feature_columns])
    test_features = scaler_X.transform(test_data[feature_columns])

    # Scale target separately
    train_target = scaler_y.fit_transform(train_data[[target_column]])
    val_target = scaler_y.transform(val_data[[target_column]])
    test_target = scaler_y.transform(test_data[[target_column]])

    # Flatten target arrays
    train_target = train_target.flatten()
    val_target = val_target.flatten()
    test_target = test_target.flatten()

    # Create datasets
    return {
        'train_features': train_features,
        'train_target': train_target,
        'val_features': val_features,
        'val_target': val_target,
        'test_features': test_features,
        'test_target': test_target,
        'feature_columns': feature_columns,
        'target_column': target_column,
        'scaler_X': scaler_X,
        'scaler_y': scaler_y,
        'train_data': train_data,
        'val_data': val_data,
        'test_data': test_data
    }

# Create data loaders
def create_dataloaders(data_dict, params):
    # Create datasets
    train_dataset = ElectricityLoadDataset(
        data_dict['train_features'],
        data_dict['train_target'],
        params.sequence_length
    )

    val_dataset = ElectricityLoadDataset(
        data_dict['val_features'],
        data_dict['val_target'],
        params.sequence_length
    )

    test_dataset = ElectricityLoadDataset(
        data_dict['test_features'],
        data_dict['test_target'],
        params.sequence_length
    )

    # Create data loaders
    train_loader = DataLoader(
        train_dataset,
        batch_size=params.batch_size,
        shuffle=True,
        num_workers=2,
        pin_memory=True if torch.cuda.is_available() else False
    )

    val_loader = DataLoader(
        val_dataset,
        batch_size=params.batch_size,
        shuffle=False,
        num_workers=2,
        pin_memory=True if torch.cuda.is_available() else False
    )

    test_loader = DataLoader(
        test_dataset,
        batch_size=params.batch_size,
        shuffle=False,
        num_workers=2,
        pin_memory=True if torch.cuda.is_available() else False
    )

    return train_loader, val_loader, test_loader

# Define the evaluation metrics
def calculate_metrics(y_true, y_pred):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)

    # MSE
    mse = mean_squared_error(y_true, y_pred)
    # RMSE
    rmse = np.sqrt(mse)
    # MAE
    mae = mean_absolute_error(y_true, y_pred)
    # NRMSE (Normalized RMSE)
    nrmse = rmse / (np.max(y_true) - np.min(y_true))
    # MAPE (Mean Absolute Percentage Error)
    # Avoid division by zero
    mask = y_true != 0
    mape = np.mean(np.abs((y_true[mask] - y_pred[mask]) / y_true[mask])) * 100

    return {
        'MSE': mse,
        'RMSE': rmse,
        'MAE': mae,
        'NRMSE': nrmse,
        'MAPE': mape
    }

# Training function
def train(model, train_loader, val_loader, optimizer, scheduler, num_epochs, patience, device):
    train_losses = []
    val_losses = []
    best_val_loss = float('inf')
    patience_counter = 0
    best_model = None

    for epoch in range(num_epochs):
        # Training
        model.train()
        train_loss = 0

        for X_batch, y_batch in train_loader:
            X_batch, y_batch = X_batch.to(device), y_batch.to(device)

            # Forward pass
            y_pred = model(X_batch)

            # Calculate loss
            loss = pinball_loss(y_pred, y_batch)

            # Backward and optimize
            optimizer.zero_grad()
            loss.backward()

            # Apply gradient clipping to prevent exploding gradients
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)

            optimizer.step()

            train_loss += loss.item()

        train_loss /= len(train_loader)
        train_losses.append(train_loss)

        # Validation
        model.eval()
        val_loss = 0

        with torch.no_grad():
            for X_batch, y_batch in val_loader:
                X_batch, y_batch = X_batch.to(device), y_batch.to(device)

                # Forward pass
                y_pred = model(X_batch)

                # Calculate loss
                loss = pinball_loss(y_pred, y_batch)
                val_loss += loss.item()

        val_loss /= len(val_loader)
        val_losses.append(val_loss)

        # Update learning rate
        scheduler.step(val_loss)

        # Print progress
        print(f'Epoch [{epoch+1}/{num_epochs}], Train Loss: {train_loss:.4f}, Val Loss: {val_loss:.4f}')

        # Check for early stopping
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            patience_counter = 0
            # Save best model
            best_model = model.state_dict().copy()
        else:
            patience_counter += 1
            if patience_counter >= patience:
                print(f'Early stopping at epoch {epoch+1}')
                break

    # Load best model
    model.load_state_dict(best_model)

    return model, train_losses, val_losses

# Evaluation function
def evaluate(model, test_loader, scaler_y, device):
    model.eval()

    all_preds = []
    all_targets = []

    with torch.no_grad():
        for X_batch, y_batch in test_loader:
            X_batch, y_batch = X_batch.to(device), y_batch.to(device)

            # Forward pass
            y_pred = model(X_batch)

            # Extract median prediction (quantile 0.5)
            median_idx = QUANTILES.index(0.5)
            y_pred_median = y_pred[:, median_idx]

            # Convert to numpy and transform back
            y_pred_np = y_pred_median.cpu().numpy()
            y_batch_np = y_batch.cpu().numpy()

            # Store predictions and targets
            all_preds.extend(y_pred_np)
            all_targets.extend(y_batch_np)

    # Convert to numpy arrays
    all_preds = np.array(all_preds).reshape(-1, 1)
    all_targets = np.array(all_targets).reshape(-1, 1)

    # Inverse transform to original scale
    all_preds = scaler_y.inverse_transform(all_preds)
    all_targets = scaler_y.inverse_transform(all_targets)

    # Calculate metrics
    metrics = calculate_metrics(all_targets, all_preds)

    return all_preds, all_targets, metrics

# Plot training and validation loss
def plot_losses(train_losses, val_losses):
    plt.figure(figsize=(10, 6))
    plt.plot(train_losses, label='Train Loss')
    plt.plot(val_losses, label='Validation Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.title('Training and Validation Loss')
    plt.legend()
    plt.grid(True)
    plt.savefig('panama_model_loss_plot.png')
    plt.close()

# Plot predictions vs actual
def plot_predictions(y_true, y_pred, title='Predictions vs Actual'):
    plt.figure(figsize=(15, 8))
    plt.plot(y_true, label='Actual', color='blue')
    plt.plot(y_pred, label='Predicted', color='red', alpha=0.7)
    plt.xlabel('Time Step')
    plt.ylabel('Electricity Load')
    plt.title(title)
    plt.legend()
    plt.grid(True)
    plt.savefig('panama_model_predictions_plot.png')
    plt.close()

    # Plot a section of the predictions for better visibility
    # Take a smaller window (e.g., 2 weeks) for more detailed visualization
    window_size = 24 * 14  # 14 days assuming hourly data
    if len(y_true) > window_size:
        start_idx = len(y_true) // 2 - window_size // 2  # Center the window
        end_idx = start_idx + window_size

        # Plot the zoomed section
        plt.figure(figsize=(15, 8))
        plt.plot(y_true[start_idx:end_idx], label='Actual', color='blue')
        plt.plot(y_pred[start_idx:end_idx], label='Predicted', color='red', alpha=0.7)
        plt.xlabel('Time Step')
        plt.ylabel('Electricity Load')
        plt.title(f'{title} (2-Week Window)')
        plt.legend()
        plt.grid(True)
        plt.savefig('panama_model_zoomed_predictions_plot.png')
        plt.close()

# Optuna objective function for hyperparameter tuning
def objective(trial, data_dict):
    # Define the hyperparameters to optimize
    params = HyperParams()
    params.batch_size = trial.suggest_categorical('batch_size', [32, 64, 128])
    params.hidden_size = trial.suggest_categorical('hidden_size', [64, 128, 256])
    params.num_layers = trial.suggest_int('num_layers', 1, 3)
    params.dropout = trial.suggest_float('dropout', 0.1, 0.5)
    params.learning_rate = trial.suggest_loguniform('learning_rate', 1e-4, 1e-2)
    params.weight_decay = trial.suggest_loguniform('weight_decay', 1e-6, 1e-4)
    params.cnn_filters = trial.suggest_categorical('cnn_filters', [32, 64, 128])
    params.cnn_kernel_size = trial.suggest_categorical('cnn_kernel_size', [3, 5, 7])
    params.nhead = trial.suggest_categorical('nhead', [4, 8])
    params.transformer_dim = trial.suggest_categorical('transformer_dim', [64, 128, 256])
    params.transformer_layers = trial.suggest_int('transformer_layers', 1, 3)
    params.local_window_size = trial.suggest_int('local_window_size', 4, 12)

    # Create data loaders
    train_loader, val_loader, _ = create_dataloaders(data_dict, params)

    # Create model
    input_dim = data_dict['train_features'].shape[1]
    output_dim = 1  # Predicting a single value

    model = CNNTransformerLSTM(input_dim, output_dim, params).to(device)

    # Define optimizer and scheduler
    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=params.learning_rate,
        weight_decay=params.weight_decay
    )

    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode='min',
        factor=0.5,
        patience=5,
        verbose=True
    )

    # Train for fewer epochs during optimization
    num_epochs = 30
    patience = 7

    # Train model
    _, train_losses, val_losses = train(
        model,
        train_loader,
        val_loader,
        optimizer,
        scheduler,
        num_epochs,
        patience,
        device
    )

    # Return the best validation loss
    return min(val_losses)

# Main function
def main():
    # Load data
    print("Loading data...")
    df = pd.read_csv('/content/drive/MyDrive/thesis Final/Implementation/panama_dataset.csv')

    print(f"Loaded dataset with {len(df)} rows and {len(df.columns)} columns")
    print(f"Columns: {df.columns.tolist()}")

    # Preprocess data
    print("Preprocessing data...")
    data_dict = preprocess_data(df)

    print(f"After preprocessing:")
    print(f"Train set: {data_dict['train_features'].shape[0]} samples")
    print(f"Validation set: {data_dict['val_features'].shape[0]} samples")
    print(f"Test set: {data_dict['test_features'].shape[0]} samples")
    print(f"Number of features: {data_dict['train_features'].shape[1]}")

    # Due to computational constraints, we'll use a smaller hyperparameter search
    # You can set this to True if you want to run the full hyperparameter search
    run_optuna = True

    if run_optuna:
        # Run Optuna optimization
        print("Running hyperparameter optimization with Optuna...")
        study = optuna.create_study(direction='minimize')
        study.optimize(lambda trial: objective(trial, data_dict), n_trials=20)

        # Get the best hyperparameters
        best_params = study.best_params
        print(f"Best hyperparameters: {best_params}")

        # Apply best hyperparameters
        params = HyperParams()
        params.batch_size = best_params['batch_size']
        params.hidden_size = best_params['hidden_size']
        params.num_layers = best_params['num_layers']
        params.dropout = best_params['dropout']
        params.learning_rate = best_params['learning_rate']
        params.weight_decay = best_params['weight_decay']
        params.cnn_filters = best_params['cnn_filters']
        params.cnn_kernel_size = best_params['cnn_kernel_size']
        params.nhead = best_params['nhead']
        params.transformer_dim = best_params['transformer_dim']
        params.transformer_layers = best_params['transformer_layers']
        params.local_window_size = best_params['local_window_size']
    else:
        # Use default hyperparameters
        print("Using default hyperparameters...")
        params = HyperParams()
        # Customize default parameters for this dataset
        params.sequence_length = 24
        params.batch_size = 64
        params.hidden_size = 128
        params.num_layers = 2
        params.dropout = 0.3
        params.learning_rate = 0.001
        params.weight_decay = 1e-5
        params.cnn_filters = 64
        params.cnn_kernel_size = 3
        params.nhead = 8
        params.transformer_dim = 128
        params.transformer_layers = 2
        params.local_window_size = 6

    # Create data loaders
    train_loader, val_loader, test_loader = create_dataloaders(data_dict, params)

    # Create model
    input_dim = data_dict['train_features'].shape[1]
    output_dim = 1  # Predicting a single value

    model = CNNTransformerLSTM(input_dim, output_dim, params).to(device)
    print(f"Model created with input dimension: {input_dim}, output dimension: {output_dim}")

    # Print model summary (number of parameters)
    total_params = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"Total parameters: {total_params}")
    print(f"Trainable parameters: {trainable_params}")

    # Define optimizer and scheduler
    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=params.learning_rate,
        weight_decay=params.weight_decay
    )

    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode='min',
        factor=0.5,
        patience=5,
        verbose=True
    )

    # Train the model
    print("Training model...")
    num_epochs = 100  # Reduced for faster execution, increase for better results
    patience = params.patience  # Early stopping patience

    final_model, train_losses, val_losses = train(
        model,
        train_loader,
        val_loader,
        optimizer,
        scheduler,
        num_epochs,
        patience,
        device
    )

    # Plot training and validation losses
    plot_losses(train_losses, val_losses)

    # Evaluate on test set
    print("Evaluating model on test set...")
    predictions, targets, metrics = evaluate(final_model, test_loader, data_dict['scaler_y'], device)

    # Print metrics
    print("\nTest Set Metrics:")
    for metric_name, metric_value in metrics.items():
        print(f"{metric_name}: {metric_value:.4f}")

    # Plot predictions vs actual
    print("Plotting predictions vs actual...")
    plot_predictions(targets.flatten(), predictions.flatten(), title='Panama Electricity Demand: Predictions vs Actual')

    # Save the model
    torch.save(final_model.state_dict(), 'panama_model.pth')
    print("Model saved successfully!")

    # Save predictions for further analysis
    test_dates = data_dict['test_data'].index[-len(predictions):]
    pred_df = pd.DataFrame({
        'datetime': test_dates,
        'actual': targets.flatten(),
        'predicted': predictions.flatten()
    })
    pred_df.to_csv('panama_predictions.csv', index=False)
    print("Predictions saved to panama_predictions.csv")

    return final_model, metrics, pred_df


if __name__ == "__main__":
    # Import seaborn for better visualizations
    try:
        import seaborn as sns
        sns.set_theme(style="whitegrid")
    except ImportError:
        print("Seaborn not installed. Using matplotlib for visualizations.")

    # Run the main function
    model, metrics, predictions_df = main()

    print("\nFinal Test Set Metrics:")
    for metric_name, metric_value in metrics.items():
        print(f"{metric_name}: {metric_value:.4f}")

    print("\nAnalysis Complete!")
    print("Model has been trained on the Panama dataset and saved as 'panama_model.pth'")
    print("Predictions have been saved to 'panama_predictions.csv'")
    print("Visualization plots have been saved as PNG files")
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error
import warnings
import math
warnings.filterwarnings('ignore')

# Set seeds for reproducibility
SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

# Device configuration
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# Quantile levels for prediction
QUANTILES = [0.1, 0.5, 0.9]

# Hyperparameters (based on provided best parameters)
class HyperParams:
    def __init__(self):
        # Using the provided best hyperparameters
        self.batch_size = 32
        self.sequence_length = 24  # As specified in the original code
        self.hidden_size = 128
        self.num_layers = 1
        self.dropout = 0.12923361001405317
        self.learning_rate = 0.000641285932448859
        self.weight_decay = 1.8179096220981544e-05
        self.cnn_filters = 32
        self.cnn_kernel_size = 5
        self.nhead = 8
        self.transformer_dim = 128
        self.transformer_layers = 3
        self.patience = 10  # Early stopping patience
        self.local_window_size = 8  # Size of local attention window

# Custom dataset class
class ElectricityLoadDataset(Dataset):
    def __init__(self, features, targets, sequence_length):
        self.features = features
        self.targets = targets
        self.sequence_length = sequence_length

    def __len__(self):
        return len(self.features) - self.sequence_length

    def __getitem__(self, idx):
        x = self.features[idx:idx + self.sequence_length]
        y = self.targets[idx + self.sequence_length]
        return torch.FloatTensor(x), torch.FloatTensor([y])

# CNN Block
class CNNBlock(nn.Module):
    def __init__(self, in_channels, out_channels, kernel_size, dropout):
        super(CNNBlock, self).__init__()
        self.conv1 = nn.Conv1d(in_channels, out_channels, kernel_size, padding=kernel_size//2)
        self.bn1 = nn.BatchNorm1d(out_channels)
        self.conv2 = nn.Conv1d(out_channels, out_channels, kernel_size, padding=kernel_size//2)
        self.bn2 = nn.BatchNorm1d(out_channels)
        self.relu = nn.ReLU()
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # x shape: [batch, sequence, features]
        x = x.permute(0, 2, 1)  # Change to [batch, features, sequence]
        x = self.relu(self.bn1(self.conv1(x)))
        x = self.dropout(x)
        x = self.relu(self.bn2(self.conv2(x)))
        x = self.dropout(x)
        x = x.permute(0, 2, 1)  # Change back to [batch, sequence, features]
        return x

# Positional Encoding for Transformer
class PositionalEncoding(nn.Module):
    def __init__(self, d_model, dropout=0.1, max_len=5000):
        super(PositionalEncoding, self).__init__()
        self.dropout = nn.Dropout(p=dropout)

        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-np.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        pe = pe.unsqueeze(0)
        self.register_buffer('pe', pe)

    def forward(self, x):
        x = x + self.pe[:, :x.size(1), :]
        return self.dropout(x)

# Transformer Block
class TransformerBlock(nn.Module):
    def __init__(self, d_model, nhead, dim_feedforward, dropout, num_layers):
        super(TransformerBlock, self).__init__()
        encoder_layers = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=nhead,
            dim_feedforward=dim_feedforward,
            dropout=dropout,
            batch_first=True  # For [batch, seq, feature] input
        )
        self.transformer_encoder = nn.TransformerEncoder(encoder_layers, num_layers=num_layers)
        self.pos_encoder = PositionalEncoding(d_model, dropout)

    def forward(self, x):
        # Add positional encoding
        x = self.pos_encoder(x)
        # Apply transformer
        return self.transformer_encoder(x)

# Local-Global Attention Module
class LocalGlobalAttention(nn.Module):
    def __init__(self, embed_dim, num_heads, local_window_size, dropout=0.1):
        super(LocalGlobalAttention, self).__init__()
        self.embed_dim = embed_dim
        self.num_heads = num_heads
        self.head_dim = embed_dim // num_heads
        self.local_window_size = local_window_size

        assert self.head_dim * num_heads == embed_dim, "embed_dim must be divisible by num_heads"

        # Local attention (focused on nearby timesteps)
        self.local_attention = nn.MultiheadAttention(
            embed_dim=embed_dim,
            num_heads=num_heads,
            dropout=dropout,
            batch_first=True
        )

        # Global attention (considering the entire sequence)
        self.global_attention = nn.MultiheadAttention(
            embed_dim=embed_dim,
            num_heads=num_heads,
            dropout=dropout,
            batch_first=True
        )

        # Gating mechanism to balance local and global information
        self.gate = nn.Sequential(
            nn.Linear(embed_dim * 2, embed_dim),
            nn.Sigmoid()
        )

        # Layer normalization for stability
        self.norm1 = nn.LayerNorm(embed_dim)
        self.norm2 = nn.LayerNorm(embed_dim)
        self.norm_final = nn.LayerNorm(embed_dim)

        # Output projection
        self.output_projection = nn.Linear(embed_dim, embed_dim)
        self.dropout = nn.Dropout(dropout)

    def _generate_local_mask(self, seq_len, local_window_size, device):
        """Generate a mask for local attention with specified window size"""
        # Initialize mask with zeros (masked positions)
        mask = torch.zeros(seq_len, seq_len, device=device)

        # Fill diagonal band with ones (visible positions)
        for i in range(seq_len):
            # Define window boundaries
            window_start = max(0, i - local_window_size // 2)
            window_end = min(seq_len, i + local_window_size // 2 + 1)
            # Set visible positions to 1
            mask[i, window_start:window_end] = 1

        return mask

    def forward(self, x):
        # x shape: [batch_size, sequence_length, embed_dim]
        batch_size, seq_len, _ = x.shape

        # Apply normalization
        normed_x = self.norm1(x)

        # Generate local attention mask
        local_mask = self._generate_local_mask(seq_len, self.local_window_size, x.device)

        # Apply local attention (with window mask)
        local_output, _ = self.local_attention(
            query=normed_x,
            key=normed_x,
            value=normed_x,
            attn_mask=local_mask,
            need_weights=False
        )

        # Apply global attention (full sequence)
        global_output, _ = self.global_attention(
            query=self.norm2(x),
            key=self.norm2(x),
            value=self.norm2(x),
            need_weights=False
        )

        # Calculate adaptive gate values to balance local and global information
        combined = torch.cat([local_output, global_output], dim=-1)
        gate_values = self.gate(combined)

        # Mix local and global features based on gate values
        mixed_output = gate_values * local_output + (1 - gate_values) * global_output

        # Apply output projection and residual connection
        output = x + self.dropout(self.output_projection(mixed_output))

        # Final normalization
        output = self.norm_final(output)

        return output

# CNN-Transformer-LSTM Model with Local-Global Attention
class CNNTransformerLSTM(nn.Module):
    def __init__(self, input_dim, output_dim, params):
        super(CNNTransformerLSTM, self).__init__()

        self.params = params

        # CNN to extract local patterns
        self.cnn = CNNBlock(
            in_channels=input_dim,
            out_channels=params.cnn_filters,
            kernel_size=params.cnn_kernel_size,
            dropout=params.dropout
        )

        # Linear layer to map CNN output to transformer dimension
        self.cnn_to_transformer = nn.Linear(params.cnn_filters, params.transformer_dim)

        # Transformer for capturing global dependencies
        self.transformer = TransformerBlock(
            d_model=params.transformer_dim,
            nhead=params.nhead,
            dim_feedforward=params.transformer_dim * 4,
            dropout=params.dropout,
            num_layers=params.transformer_layers
        )

        # LSTM for temporal dependencies
        self.lstm = nn.LSTM(
            input_size=params.transformer_dim,
            hidden_size=params.hidden_size,
            num_layers=params.num_layers,
            batch_first=True,
            dropout=params.dropout if params.num_layers > 1 else 0,
            bidirectional=True
        )

        # Replace standard attention with Local-Global Attention
        self.local_global_attention = LocalGlobalAttention(
            embed_dim=params.hidden_size * 2,  # * 2 for bidirectional LSTM
            num_heads=params.nhead,
            local_window_size=params.local_window_size,
            dropout=params.dropout
        )

        # Dropout layer
        self.dropout = nn.Dropout(params.dropout)

        # Output layers for each quantile
        self.output_layers = nn.ModuleList([
            nn.Linear(params.hidden_size * 2, output_dim)  # * 2 for bidirectional
            for _ in QUANTILES
        ])

    def forward(self, x):
        # Apply CNN
        x_cnn = self.cnn(x)

        # Map to transformer dimension
        x_transformer_input = self.cnn_to_transformer(x_cnn)

        # Apply Transformer
        x_transformer = self.transformer(x_transformer_input)

        # Apply LSTM
        x_lstm, (h_n, c_n) = self.lstm(x_transformer)

        # Apply Local-Global attention to LSTM outputs
        x_attn = self.local_global_attention(x_lstm)

        # Get the last time step for prediction
        x_last = x_attn[:, -1, :]

        # Apply dropout
        x_last = self.dropout(x_last)

        # Apply output layers for each quantile
        quantile_outputs = [output_layer(x_last) for output_layer in self.output_layers]

        # Stack quantile outputs
        return torch.stack(quantile_outputs, dim=1)

# Pinball Loss for Quantile Regression
def pinball_loss(y_pred, y_true, quantiles=QUANTILES):
    losses = []
    for i, q in enumerate(quantiles):
        errors = y_true - y_pred[:, i]
        losses.append(torch.max(q * errors, (q - 1) * errors))

    return torch.mean(torch.stack(losses))

# Function to preprocess data
def preprocess_data(df, test_size=0.15, val_size=0.15):
    # Extract target variable
    target_column = 'target_variable'
    feature_columns = [col for col in df.columns if col != target_column]

    # Create time features assuming data is sequential
    df['hour'] = np.arange(len(df)) % 24
    df['day'] = (np.arange(len(df)) // 24) % 7
    df['month'] = (np.arange(len(df)) // (24 * 30)) % 12 + 1  # Approximation

    # Add sine and cosine transformations for cyclical features
    for col in ['hour', 'day', 'month']:
        if col == 'hour':
            max_val = 24
        elif col == 'day':
            max_val = 7
        else:  # month
            max_val = 12

        df[f'{col}_sin'] = np.sin(2 * np.pi * df[col] / max_val)
        df[f'{col}_cos'] = np.cos(2 * np.pi * df[col] / max_val)

    # Add these new columns to feature columns and remove original time columns
    feature_columns.extend([f'{col}_sin' for col in ['hour', 'day', 'month']])
    feature_columns.extend([f'{col}_cos' for col in ['hour', 'day', 'month']])

    # Create lags for the target variable
    for lag in [1, 2, 3, 6, 12, 24, 48, 72, 168]:  # 1h, 2h, 3h, 6h, 12h, 1d, 2d, 3d, 1w
        df[f'target_lag_{lag}'] = df[target_column].shift(lag)
        feature_columns.append(f'target_lag_{lag}')

    # Create rolling statistics
    for window in [6, 12, 24, 48, 72, 168]:  # 6h, 12h, 1d, 2d, 3d, 1w
        df[f'target_rolling_mean_{window}'] = df[target_column].rolling(window=window).mean().shift(1)
        df[f'target_rolling_std_{window}'] = df[target_column].rolling(window=window).std().shift(1)
        df[f'target_rolling_max_{window}'] = df[target_column].rolling(window=window).max().shift(1)
        df[f'target_rolling_min_{window}'] = df[target_column].rolling(window=window).min().shift(1)

        feature_columns.extend([
            f'target_rolling_mean_{window}',
            f'target_rolling_std_{window}',
            f'target_rolling_max_{window}',
            f'target_rolling_min_{window}'
        ])

    # Drop rows with NaN values (due to lagging and rolling operations)
    df.dropna(inplace=True)

    # Split data based on chronological order (since no year column in the dataset)
    total_samples = len(df)
    test_samples = int(test_size * total_samples)
    val_samples = int(val_size * total_samples)
    train_samples = total_samples - test_samples - val_samples

    train_data = df.iloc[:train_samples]
    val_data = df.iloc[train_samples:train_samples+val_samples]
    test_data = df.iloc[train_samples+val_samples:]

    # Scale features
    scaler_X = StandardScaler()
    scaler_y = StandardScaler()

    # Fit on training data
    train_features = scaler_X.fit_transform(train_data[feature_columns])
    val_features = scaler_X.transform(val_data[feature_columns])
    test_features = scaler_X.transform(test_data[feature_columns])

    # Scale target separately
    train_target = scaler_y.fit_transform(train_data[[target_column]])
    val_target = scaler_y.transform(val_data[[target_column]])
    test_target = scaler_y.transform(test_data[[target_column]])

    # Flatten target arrays
    train_target = train_target.flatten()
    val_target = val_target.flatten()
    test_target = test_target.flatten()

    # Create datasets
    return {
        'train_features': train_features,
        'train_target': train_target,
        'val_features': val_features,
        'val_target': val_target,
        'test_features': test_features,
        'test_target': test_target,
        'feature_columns': feature_columns,
        'target_column': target_column,
        'scaler_X': scaler_X,
        'scaler_y': scaler_y,
        'train_data': train_data,
        'val_data': val_data,
        'test_data': test_data
    }

# Create data loaders
def create_dataloaders(data_dict, params):
    # Create datasets
    train_dataset = ElectricityLoadDataset(
        data_dict['train_features'],
        data_dict['train_target'],
        params.sequence_length
    )

    val_dataset = ElectricityLoadDataset(
        data_dict['val_features'],
        data_dict['val_target'],
        params.sequence_length
    )

    test_dataset = ElectricityLoadDataset(
        data_dict['test_features'],
        data_dict['test_target'],
        params.sequence_length
    )

    # Create data loaders
    train_loader = DataLoader(
        train_dataset,
        batch_size=params.batch_size,
        shuffle=True,
        num_workers=4,
        pin_memory=True
    )

    val_loader = DataLoader(
        val_dataset,
        batch_size=params.batch_size,
        shuffle=False,
        num_workers=4,
        pin_memory=True
    )

    test_loader = DataLoader(
        test_dataset,
        batch_size=params.batch_size,
        shuffle=False,
        num_workers=4,
        pin_memory=True
    )

    return train_loader, val_loader, test_loader

# Define the evaluation metrics
def calculate_metrics(y_true, y_pred):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)

    # MSE
    mse = mean_squared_error(y_true, y_pred)
    # RMSE
    rmse = np.sqrt(mse)
    # MAE
    mae = mean_absolute_error(y_true, y_pred)
    # NRMSE (Normalized RMSE)
    nrmse = rmse / (np.max(y_true) - np.min(y_true))
    # MAPE (Mean Absolute Percentage Error)
    mape = np.mean(np.abs((y_true - y_pred) / y_true)) * 100

    return {
        'MSE': mse,
        'RMSE': rmse,
        'MAE': mae,
        'NRMSE': nrmse,
        'MAPE': mape
    }

# Training function
def train(model, train_loader, val_loader, optimizer, scheduler, num_epochs, patience, device):
    train_losses = []
    val_losses = []
    best_val_loss = float('inf')
    patience_counter = 0
    best_model = None

    for epoch in range(num_epochs):
        # Training
        model.train()
        train_loss = 0

        for X_batch, y_batch in train_loader:
            X_batch, y_batch = X_batch.to(device), y_batch.to(device)

            # Forward pass
            y_pred = model(X_batch)

            # Calculate loss
            loss = pinball_loss(y_pred, y_batch)

            # Backward and optimize
            optimizer.zero_grad()
            loss.backward()

            # Apply gradient clipping to prevent exploding gradients
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)

            optimizer.step()

            train_loss += loss.item()

        train_loss /= len(train_loader)
        train_losses.append(train_loss)

        # Validation
        model.eval()
        val_loss = 0

        with torch.no_grad():
            for X_batch, y_batch in val_loader:
                X_batch, y_batch = X_batch.to(device), y_batch.to(device)

                # Forward pass
                y_pred = model(X_batch)

                # Calculate loss
                loss = pinball_loss(y_pred, y_batch)
                val_loss += loss.item()

        val_loss /= len(val_loader)
        val_losses.append(val_loss)

        # Update learning rate
        scheduler.step(val_loss)

        # Print progress
        print(f'Epoch [{epoch+1}/{num_epochs}], Train Loss: {train_loss:.4f}, Val Loss: {val_loss:.4f}')

        # Check for early stopping
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            patience_counter = 0
            # Save best model
            best_model = model.state_dict().copy()
        else:
            patience_counter += 1
            if patience_counter >= patience:
                print(f'Early stopping at epoch {epoch+1}')
                break

    # Load best model
    model.load_state_dict(best_model)

    return model, train_losses, val_losses

# Evaluation function
def evaluate(model, test_loader, scaler_y, device):
    model.eval()

    all_preds = []
    all_targets = []

    with torch.no_grad():
        for X_batch, y_batch in test_loader:
            X_batch, y_batch = X_batch.to(device), y_batch.to(device)

            # Forward pass
            y_pred = model(X_batch)

            # Extract median prediction (quantile 0.5)
            median_idx = QUANTILES.index(0.5)
            y_pred_median = y_pred[:, median_idx]

            # Convert to numpy and transform back
            y_pred_np = y_pred_median.cpu().numpy()
            y_batch_np = y_batch.cpu().numpy()

            # Store predictions and targets
            all_preds.extend(y_pred_np)
            all_targets.extend(y_batch_np)

    # Convert to numpy arrays
    all_preds = np.array(all_preds).reshape(-1, 1)
    all_targets = np.array(all_targets).reshape(-1, 1)

    # Inverse transform to original scale
    all_preds = scaler_y.inverse_transform(all_preds)
    all_targets = scaler_y.inverse_transform(all_targets)

    # Calculate metrics
    metrics = calculate_metrics(all_targets, all_preds)

    return all_preds, all_targets, metrics

# Plot training and validation loss
def plot_losses(train_losses, val_losses):
    plt.figure(figsize=(10, 6))
    plt.plot(train_losses, label='Train Loss')
    plt.plot(val_losses, label='Validation Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.title('Training and Validation Loss')
    plt.legend()
    plt.grid(True)
    plt.savefig('local_global_attention_loss_plot.png')
    plt.show()

# Plot predictions vs actual
def plot_predictions(y_true, y_pred, title='Predictions vs Actual'):
    plt.figure(figsize=(15, 8))
    plt.plot(y_true, label='Actual', color='blue')
    plt.plot(y_pred, label='Predicted', color='red', alpha=0.7)
    plt.xlabel('Time Step')
    plt.ylabel('Target Variable')
    plt.title(title)
    plt.legend()
    plt.grid(True)
    plt.savefig('local_global_attention_predictions_plot.png')
    plt.show()

# Main function
def main():
    # Load and preprocess data
    print("Loading and preprocessing data...")
    df = pd.read_csv('/content/drive/MyDrive/thesis Final/Implementation/panama_dataset.csv')

    # Preprocess data
    data_dict = preprocess_data(df)

    # Create data loaders
    params = HyperParams()
    train_loader, val_loader, test_loader = create_dataloaders(data_dict, params)

    # Create model with best parameters
    input_dim = data_dict['train_features'].shape[1]
    output_dim = 1  # Predicting a single value

    model = CNNTransformerLSTM(input_dim, output_dim, params).to(device)

    # Define optimizer and scheduler
    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=params.learning_rate,
        weight_decay=params.weight_decay
    )

    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode='min',
        factor=0.5,
        patience=10,
        verbose=True
    )

    # Train the final model
    print("Training model with best hyperparameters...")
    num_epochs = 100
    patience = params.patience

    final_model, train_losses, val_losses = train(
        model,
        train_loader,
        val_loader,
        optimizer,
        scheduler,
        num_epochs,
        patience,
        device
    )

    # Plot training and validation losses
    plot_losses(train_losses, val_losses)

    # Evaluate on test set
    print("Evaluating model on test set...")
    predictions, targets, metrics = evaluate(final_model, test_loader, data_dict['scaler_y'], device)

    # Print metrics
    print("\nTest Set Metrics:")
    for metric_name, metric_value in metrics.items():
        print(f"{metric_name}: {metric_value:.4f}")

    # Plot predictions vs actual
    print("Plotting predictions vs actual...")
    plot_predictions(targets.flatten(), predictions.flatten(), title='CNN-Transformer-LSTM with Local-Global Attention: Predictions vs Actual')

    # Plot a section of the predictions for better visibility
    window_size = 24 * 14  # 14 days assuming hourly data
    if len(targets) > window_size:
        start_idx = len(targets) // 2 - window_size // 2  # Center the window
        end_idx = start_idx + window_size

        # Plot the zoomed section
        plt.figure(figsize=(15, 8))
        plt.plot(targets[start_idx:end_idx], label='Actual', color='blue')
        plt.plot(predictions[start_idx:end_idx], label='Predicted', color='red', alpha=0.7)
        plt.xlabel('Time Step')
        plt.ylabel('Target Variable')
        plt.title('CNN-Transformer-LSTM with Local-Global Attention: Predictions vs Actual (2-Week Window)')
        plt.legend()
        plt.grid(True)
        plt.savefig('local_global_attention_zoomed_predictions_plot.png')
        plt.show()

    # Save the model
    torch.save(final_model.state_dict(), 'panama_model.pth')
    print("Model saved successfully!")

    return final_model, metrics


if __name__ == "__main__":
    # Try importing seaborn for better visualizations
    try:
        import seaborn as sns
        sns.set_theme(style="whitegrid")
    except ImportError:
        print("Seaborn not installed. Using matplotlib for visualizations.")

    # Load the data
    df = pd.read_csv('/content/drive/MyDrive/thesis Final/Implementation/panama_dataset.csv')

    # Run the main function
    model, metrics = main()

    print("\nFinal Test Set Metrics:")
    for metric_name, metric_value in metrics.items():
        print(f"{metric_name}: {metric_value:.4f}")
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error
import warnings
import math
import optuna  # Import Optuna for hyperparameter tuning
from optuna.pruners import MedianPruner  # Import pruner for early stopping inefficient trials
import copy
warnings.filterwarnings('ignore')

# Set seeds for reproducibility
SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

# Device configuration
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# Quantile levels for prediction
QUANTILES = [0.1, 0.5, 0.9]

# Class for hyperparameters (defaults will be used for preprocessing only)
class HyperParams:
    def __init__(self):
        # Default hyperparameters (will be tuned by Optuna)
        self.batch_size = 32
        self.sequence_length = 24  # As specified in the requirements
        self.hidden_size = 128
        self.num_layers = 1
        self.dropout = 0.1
        self.learning_rate = 0.001
        self.weight_decay = 1e-5
        self.cnn_filters = 32
        self.cnn_kernel_size = 5
        self.nhead = 8
        self.transformer_dim = 128
        self.transformer_layers = 2
        self.patience = 10  # Early stopping patience
        self.local_window_size = 8  # Size of local attention window (in timesteps)

# Custom dataset class
class ElectricityLoadDataset(Dataset):
    def __init__(self, features, targets, sequence_length):
        self.features = features
        self.targets = targets
        self.sequence_length = sequence_length

    def __len__(self):
        return len(self.features) - self.sequence_length

    def __getitem__(self, idx):
        x = self.features[idx:idx + self.sequence_length]
        y = self.targets[idx + self.sequence_length]
        return torch.FloatTensor(x), torch.FloatTensor([y])

# CNN Block
class CNNBlock(nn.Module):
    def __init__(self, in_channels, out_channels, kernel_size, dropout):
        super(CNNBlock, self).__init__()
        self.conv1 = nn.Conv1d(in_channels, out_channels, kernel_size, padding=kernel_size//2)
        self.bn1 = nn.BatchNorm1d(out_channels)
        self.conv2 = nn.Conv1d(out_channels, out_channels, kernel_size, padding=kernel_size//2)
        self.bn2 = nn.BatchNorm1d(out_channels)
        self.relu = nn.ReLU()
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # x shape: [batch, sequence, features]
        x = x.permute(0, 2, 1)  # Change to [batch, features, sequence]
        x = self.relu(self.bn1(self.conv1(x)))
        x = self.dropout(x)
        x = self.relu(self.bn2(self.conv2(x)))
        x = self.dropout(x)
        x = x.permute(0, 2, 1)  # Change back to [batch, sequence, features]
        return x

# Positional Encoding for Transformer
class PositionalEncoding(nn.Module):
    def __init__(self, d_model, dropout=0.1, max_len=5000):
        super(PositionalEncoding, self).__init__()
        self.dropout = nn.Dropout(p=dropout)

        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-np.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        pe = pe.unsqueeze(0)
        self.register_buffer('pe', pe)

    def forward(self, x):
        x = x + self.pe[:, :x.size(1), :]
        return self.dropout(x)

# Transformer Block
class TransformerBlock(nn.Module):
    def __init__(self, d_model, nhead, dim_feedforward, dropout, num_layers):
        super(TransformerBlock, self).__init__()
        encoder_layers = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=nhead,
            dim_feedforward=dim_feedforward,
            dropout=dropout,
            batch_first=True  # For [batch, seq, feature] input
        )
        self.transformer_encoder = nn.TransformerEncoder(encoder_layers, num_layers=num_layers)
        self.pos_encoder = PositionalEncoding(d_model, dropout)

    def forward(self, x):
        # Add positional encoding
        x = self.pos_encoder(x)
        # Apply transformer
        return self.transformer_encoder(x)

# Local-Global Attention Module
class LocalGlobalAttention(nn.Module):
    def __init__(self, embed_dim, num_heads, local_window_size, dropout=0.1):
        super(LocalGlobalAttention, self).__init__()
        self.embed_dim = embed_dim
        self.num_heads = num_heads
        self.head_dim = embed_dim // num_heads
        self.local_window_size = local_window_size

        assert self.head_dim * num_heads == embed_dim, "embed_dim must be divisible by num_heads"

        # Local attention (focused on nearby timesteps)
        self.local_attention = nn.MultiheadAttention(
            embed_dim=embed_dim,
            num_heads=num_heads,
            dropout=dropout,
            batch_first=True
        )

        # Global attention (considering the entire sequence)
        self.global_attention = nn.MultiheadAttention(
            embed_dim=embed_dim,
            num_heads=num_heads,
            dropout=dropout,
            batch_first=True
        )

        # Gating mechanism to balance local and global information
        self.gate = nn.Sequential(
            nn.Linear(embed_dim * 2, embed_dim),
            nn.Sigmoid()
        )

        # Layer normalization for stability
        self.norm1 = nn.LayerNorm(embed_dim)
        self.norm2 = nn.LayerNorm(embed_dim)
        self.norm_final = nn.LayerNorm(embed_dim)

        # Output projection
        self.output_projection = nn.Linear(embed_dim, embed_dim)
        self.dropout = nn.Dropout(dropout)

    def _generate_local_mask(self, seq_len, local_window_size, device):
        """Generate a mask for local attention with specified window size"""
        # Initialize mask with zeros (masked positions)
        mask = torch.zeros(seq_len, seq_len, device=device)

        # Fill diagonal band with ones (visible positions)
        for i in range(seq_len):
            # Define window boundaries
            window_start = max(0, i - local_window_size // 2)
            window_end = min(seq_len, i + local_window_size // 2 + 1)
            # Set visible positions to 1
            mask[i, window_start:window_end] = 1

        return mask

    def forward(self, x):
        # x shape: [batch_size, sequence_length, embed_dim]
        batch_size, seq_len, _ = x.shape

        # Apply normalization
        normed_x = self.norm1(x)

        # Generate local attention mask
        local_mask = self._generate_local_mask(seq_len, self.local_window_size, x.device)

        # Apply local attention (with window mask)
        local_output, _ = self.local_attention(
            query=normed_x,
            key=normed_x,
            value=normed_x,
            attn_mask=local_mask,
            need_weights=False
        )

        # Apply global attention (full sequence)
        global_output, _ = self.global_attention(
            query=self.norm2(x),
            key=self.norm2(x),
            value=self.norm2(x),
            need_weights=False
        )

        # Calculate adaptive gate values to balance local and global information
        combined = torch.cat([local_output, global_output], dim=-1)
        gate_values = self.gate(combined)

        # Mix local and global features based on gate values
        mixed_output = gate_values * local_output + (1 - gate_values) * global_output

        # Apply output projection and residual connection
        output = x + self.dropout(self.output_projection(mixed_output))

        # Final normalization
        output = self.norm_final(output)

        return output

# CNN-Transformer-LSTM Model with Local-Global Attention
class CNNTransformerLSTM(nn.Module):
    def __init__(self, input_dim, output_dim, params):
        super(CNNTransformerLSTM, self).__init__()

        self.params = params

        # CNN to extract local patterns
        self.cnn = CNNBlock(
            in_channels=input_dim,
            out_channels=params.cnn_filters,
            kernel_size=params.cnn_kernel_size,
            dropout=params.dropout
        )

        # Linear layer to map CNN output to transformer dimension
        self.cnn_to_transformer = nn.Linear(params.cnn_filters, params.transformer_dim)

        # Transformer for capturing global dependencies
        self.transformer = TransformerBlock(
            d_model=params.transformer_dim,
            nhead=params.nhead,
            dim_feedforward=params.transformer_dim * 4,
            dropout=params.dropout,
            num_layers=params.transformer_layers
        )

        # LSTM for temporal dependencies
        self.lstm = nn.LSTM(
            input_size=params.transformer_dim,
            hidden_size=params.hidden_size,
            num_layers=params.num_layers,
            batch_first=True,
            dropout=params.dropout if params.num_layers > 1 else 0,
            bidirectional=True
        )

        # Replace standard attention with Local-Global Attention
        self.local_global_attention = LocalGlobalAttention(
            embed_dim=params.hidden_size * 2,  # * 2 for bidirectional LSTM
            num_heads=params.nhead,
            local_window_size=params.local_window_size,
            dropout=params.dropout
        )

        # Dropout layer
        self.dropout = nn.Dropout(params.dropout)

        # Output layers for each quantile
        self.output_layers = nn.ModuleList([
            nn.Linear(params.hidden_size * 2, output_dim)  # * 2 for bidirectional
            for _ in QUANTILES
        ])

    def forward(self, x):
        # Apply CNN
        x_cnn = self.cnn(x)

        # Map to transformer dimension
        x_transformer_input = self.cnn_to_transformer(x_cnn)

        # Apply Transformer
        x_transformer = self.transformer(x_transformer_input)

        # Apply LSTM
        x_lstm, (h_n, c_n) = self.lstm(x_transformer)

        # Apply Local-Global attention to LSTM outputs
        x_attn = self.local_global_attention(x_lstm)

        # Get the last time step for prediction
        x_last = x_attn[:, -1, :]

        # Apply dropout
        x_last = self.dropout(x_last)

        # Apply output layers for each quantile
        quantile_outputs = [output_layer(x_last) for output_layer in self.output_layers]

        # Stack quantile outputs
        return torch.stack(quantile_outputs, dim=1)

# Pinball Loss for Quantile Regression
def pinball_loss(y_pred, y_true, quantiles=QUANTILES):
    losses = []
    for i, q in enumerate(quantiles):
        errors = y_true - y_pred[:, i]
        losses.append(torch.max(q * errors, (q - 1) * errors))

    return torch.mean(torch.stack(losses))

# Function to preprocess the Albania dataset
def preprocess_albania_data(df):
    # Convert Time to datetime
    df['datetime'] = pd.to_datetime(df['Time'])
    df.set_index('datetime', inplace=True)

    # Extract year for train/test split
    df['year'] = df.index.year

    # Handle missing values (if any)
    df.fillna(method='ffill', inplace=True)
    df.fillna(method='bfill', inplace=True)  # In case there are missing values at the beginning

    # Extract features and target
    target_column = 'Load'
    # Exclude 'Time', 'Time_double', 'year', and target from features
    feature_columns = [col for col in df.columns if col != target_column
                      and col != 'Time' and col != 'Time_double'
                      and col != 'year']

    # Create time features
    df['hour'] = df.index.hour
    df['month'] = df.index.month
    df['day'] = df.index.day
    df['dayofweek'] = df.index.dayofweek

    # Add sine and cosine transformations for cyclical features
    for col in ['hour', 'month', 'day', 'dayofweek']:
        if col == 'hour':
            max_val = 24
        elif col == 'month':
            max_val = 12
        elif col == 'dayofweek':
            max_val = 7
        else:  # day
            max_val = 31

        df[f'{col}_sin'] = np.sin(2 * np.pi * df[col] / max_val)
        df[f'{col}_cos'] = np.cos(2 * np.pi * df[col] / max_val)

        # Remove original columns if they're in feature_columns
        if col in feature_columns:
            feature_columns.remove(col)

        # Add cyclical features to feature columns
        feature_columns.extend([f'{col}_sin', f'{col}_cos'])

    # Create lags for the target variable
    for lag in [1, 2, 3, 6, 12, 24, 48, 72, 168]:  # 1h, 2h, 3h, 6h, 12h, 1d, 2d, 3d, 1w
        df[f'load_lag_{lag}'] = df[target_column].shift(lag)
        feature_columns.append(f'load_lag_{lag}')

    # Create rolling statistics
    for window in [6, 12, 24, 48, 72, 168]:  # 6h, 12h, 1d, 2d, 3d, 1w
        df[f'load_rolling_mean_{window}'] = df[target_column].rolling(window=window).mean().shift(1)
        df[f'load_rolling_std_{window}'] = df[target_column].rolling(window=window).std().shift(1)
        df[f'load_rolling_max_{window}'] = df[target_column].rolling(window=window).max().shift(1)
        df[f'load_rolling_min_{window}'] = df[target_column].rolling(window=window).min().shift(1)

        feature_columns.extend([
            f'load_rolling_mean_{window}',
            f'load_rolling_std_{window}',
            f'load_rolling_max_{window}',
            f'load_rolling_min_{window}'
        ])

    # Drop rows with NaN values (due to lagging and rolling operations)
    df.dropna(inplace=True)

    # Split data based on year: 2016-2018 as train, rest as test
    train_years = [2016, 2017, 2018]
    train_data = df[df['year'].isin(train_years)]
    test_data = df[~df['year'].isin(train_years)]

    # Split train data into train and validation (80/20 split)
    train_indices = train_data.index
    train_val_split = int(0.8 * len(train_indices))

    # Sort the indices to ensure chronological order
    train_indices = sorted(train_indices)

    train_data_final = train_data.loc[train_indices[:train_val_split]]
    val_data = train_data.loc[train_indices[train_val_split:]]

    # Remove 'year' column
    train_data_final = train_data_final.drop('year', axis=1)
    val_data = val_data.drop('year', axis=1)
    test_data = test_data.drop('year', axis=1)

    # Scale features
    scaler_X = StandardScaler()
    scaler_y = StandardScaler()

    # Fit on training data
    train_features = scaler_X.fit_transform(train_data_final[feature_columns])
    val_features = scaler_X.transform(val_data[feature_columns])
    test_features = scaler_X.transform(test_data[feature_columns])

    # Scale target separately
    train_target = scaler_y.fit_transform(train_data_final[[target_column]])
    val_target = scaler_y.transform(val_data[[target_column]])
    test_target = scaler_y.transform(test_data[[target_column]])

    # Flatten target arrays
    train_target = train_target.flatten()
    val_target = val_target.flatten()
    test_target = test_target.flatten()

    print(f"Train data shape: {train_features.shape}, {train_target.shape}")
    print(f"Validation data shape: {val_features.shape}, {val_target.shape}")
    print(f"Test data shape: {test_features.shape}, {test_target.shape}")

    # Create datasets
    return {
        'train_features': train_features,
        'train_target': train_target,
        'val_features': val_features,
        'val_target': val_target,
        'test_features': test_features,
        'test_target': test_target,
        'feature_columns': feature_columns,
        'target_column': target_column,
        'scaler_X': scaler_X,
        'scaler_y': scaler_y,
        'train_data': train_data_final,
        'val_data': val_data,
        'test_data': test_data
    }

# Create data loaders
def create_dataloaders(data_dict, params):
    # Create datasets
    train_dataset = ElectricityLoadDataset(
        data_dict['train_features'],
        data_dict['train_target'],
        params.sequence_length
    )

    val_dataset = ElectricityLoadDataset(
        data_dict['val_features'],
        data_dict['val_target'],
        params.sequence_length
    )

    test_dataset = ElectricityLoadDataset(
        data_dict['test_features'],
        data_dict['test_target'],
        params.sequence_length
    )

    # Create data loaders
    train_loader = DataLoader(
        train_dataset,
        batch_size=params.batch_size,
        shuffle=True,
        num_workers=4,
        pin_memory=True
    )

    val_loader = DataLoader(
        val_dataset,
        batch_size=params.batch_size,
        shuffle=False,
        num_workers=4,
        pin_memory=True
    )

    test_loader = DataLoader(
        test_dataset,
        batch_size=params.batch_size,
        shuffle=False,
        num_workers=4,
        pin_memory=True
    )

    return train_loader, val_loader, test_loader

# Define the evaluation metrics
def calculate_metrics(y_true, y_pred):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)

    # MSE
    mse = mean_squared_error(y_true, y_pred)
    # RMSE
    rmse = np.sqrt(mse)
    # MAE
    mae = mean_absolute_error(y_true, y_pred)
    # NRMSE (Normalized RMSE)
    nrmse = rmse / (np.max(y_true) - np.min(y_true))
    # MAPE (Mean Absolute Percentage Error)
    # Add small epsilon to avoid division by zero
    epsilon = 1e-8
    mape = np.mean(np.abs((y_true - y_pred) / (np.abs(y_true) + epsilon))) * 100

    return {
        'MSE': mse,
        'RMSE': rmse,
        'MAE': mae,
        'NRMSE': nrmse,
        'MAPE': mape
    }

# Training function for a model with specific hyperparameters
def train_model(model, train_loader, val_loader, optimizer, scheduler, num_epochs, patience, device, trial=None):
    train_losses = []
    val_losses = []
    best_val_loss = float('inf')
    patience_counter = 0
    best_model = None

    for epoch in range(num_epochs):
        # Training
        model.train()
        train_loss = 0

        for X_batch, y_batch in train_loader:
            X_batch, y_batch = X_batch.to(device), y_batch.to(device)

            # Forward pass
            y_pred = model(X_batch)

            # Calculate loss
            loss = pinball_loss(y_pred, y_batch)

            # Backward and optimize
            optimizer.zero_grad()
            loss.backward()

            # Apply gradient clipping to prevent exploding gradients
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)

            optimizer.step()

            train_loss += loss.item()

        train_loss /= len(train_loader)
        train_losses.append(train_loss)

        # Validation
        model.eval()
        val_loss = 0

        with torch.no_grad():
            for X_batch, y_batch in val_loader:
                X_batch, y_batch = X_batch.to(device), y_batch.to(device)

                # Forward pass
                y_pred = model(X_batch)

                # Calculate loss
                loss = pinball_loss(y_pred, y_batch)
                val_loss += loss.item()

        val_loss /= len(val_loader)
        val_losses.append(val_loss)

        # Update learning rate
        scheduler.step(val_loss)

        # Print progress
        if trial is None or epoch % 5 == 0:  # Only print every 5 epochs during hyperparameter tuning
            print(f'Epoch [{epoch+1}/{num_epochs}], Train Loss: {train_loss:.4f}, Val Loss: {val_loss:.4f}')

        # Report to Optuna for pruning if in a trial
        if trial is not None:
            trial.report(val_loss, epoch)

            # Handle pruning based on the intermediate value if trial
            if trial.should_prune():
                raise optuna.exceptions.TrialPruned()

        # Check for early stopping
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            patience_counter = 0
            # Save best model
            best_model = model.state_dict().copy()
        else:
            patience_counter += 1
            if patience_counter >= patience:
                print(f'Early stopping at epoch {epoch+1}')
                break

    # Load best model
    model.load_state_dict(best_model)

    return model, train_losses, val_losses, best_val_loss

# Evaluation function
def evaluate(model, test_loader, scaler_y, device):
    model.eval()

    all_preds = []
    all_targets = []

    with torch.no_grad():
        for X_batch, y_batch in test_loader:
            X_batch, y_batch = X_batch.to(device), y_batch.to(device)

            # Forward pass
            y_pred = model(X_batch)

            # Extract median prediction (quantile 0.5)
            median_idx = QUANTILES.index(0.5)
            y_pred_median = y_pred[:, median_idx]

            # Convert to numpy and transform back
            y_pred_np = y_pred_median.cpu().numpy()
            y_batch_np = y_batch.cpu().numpy()

            # Store predictions and targets
            all_preds.extend(y_pred_np)
            all_targets.extend(y_batch_np)

    # Convert to numpy arrays
    all_preds = np.array(all_preds).reshape(-1, 1)
    all_targets = np.array(all_targets).reshape(-1, 1)

    # Inverse transform to original scale
    all_preds = scaler_y.inverse_transform(all_preds)
    all_targets = scaler_y.inverse_transform(all_targets)

    # Calculate metrics
    metrics = calculate_metrics(all_targets, all_preds)

    return all_preds, all_targets, metrics

# Plot training and validation loss
def plot_losses(train_losses, val_losses):
    plt.figure(figsize=(10, 6))
    plt.plot(train_losses, label='Train Loss')
    plt.plot(val_losses, label='Validation Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.title('Training and Validation Loss')
    plt.legend()
    plt.grid(True)
    plt.savefig('albania_optuna_loss_plot.png')
    plt.show()

# Plot predictions vs actual
def plot_predictions(y_true, y_pred, title='Predictions vs Actual'):
    plt.figure(figsize=(15, 8))
    plt.plot(y_true, label='Actual', color='blue')
    plt.plot(y_pred, label='Predicted', color='red', alpha=0.7)
    plt.xlabel('Time Step')
    plt.ylabel('Electricity Load (MWh)')
    plt.title(title)
    plt.legend()
    plt.grid(True)
    plt.savefig('albania_optuna_predictions_plot.png')
    plt.show()

# Objective function for Optuna
def objective(trial, data_dict):
    # Define the hyperparameter search space
    params = HyperParams()

    # Hyperparameters to be tuned
    params.batch_size = trial.suggest_categorical('batch_size', [16, 32, 64])
    params.hidden_size = trial.suggest_categorical('hidden_size', [64, 128, 256])
    params.num_layers = trial.suggest_int('num_layers', 1, 3)
    params.dropout = trial.suggest_float('dropout', 0.05, 0.5)
    params.learning_rate = trial.suggest_float('learning_rate', 1e-4, 1e-2, log=True)
    params.weight_decay = trial.suggest_float('weight_decay', 1e-6, 1e-4, log=True)
    params.cnn_filters = trial.suggest_categorical('cnn_filters', [16, 32, 64])
    params.cnn_kernel_size = trial.suggest_int('cnn_kernel_size', 3, 7, step=2)  # Odd numbers only
    params.nhead = trial.suggest_categorical('nhead', [4, 8, 16])
    params.transformer_dim = trial.suggest_categorical('transformer_dim', [64, 128, 256])
    params.transformer_layers = trial.suggest_int('transformer_layers', 1, 4)
    params.local_window_size = trial.suggest_int('local_window_size', 4, 12, step=2)  # Even numbers

    # Early stopping patience - not tuned but fixed
    params.patience = 10

    # Create data loaders
    train_loader, val_loader, _ = create_dataloaders(data_dict, params)

    # Setup model
    input_dim = data_dict['train_features'].shape[1]
    output_dim = 1

    model = CNNTransformerLSTM(input_dim, output_dim, params).to(device)

    # Optimizer and scheduler
    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=params.learning_rate,
        weight_decay=params.weight_decay
    )

    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode='min',
        factor=0.5,
        patience=5,
        verbose=False  # Set to False to reduce output during tuning
    )

    # Train for fewer epochs during hyperparameter search
    num_epochs = 30

    # Train the model
    try:
        _, _, _, best_val_loss = train_model(
            model,
            train_loader,
            val_loader,
            optimizer,
            scheduler,
            num_epochs,
            params.patience,
            device,
            trial=trial
        )
    except Exception as e:
        # Log exceptions during training
        print(f"Exception during training: {e}")
        return float('inf')

    return best_val_loss

# Function to run hyperparameter optimization
def run_hyperparameter_optimization(data_dict, n_trials=50):
    print("Starting hyperparameter optimization...")

    # Create a study object with pruning
    study = optuna.create_study(
        direction="minimize",
        pruner=MedianPruner(n_startup_trials=5, n_warmup_steps=5, interval_steps=1)
    )

    # Optimize with the objective function
    study.optimize(lambda trial: objective(trial, data_dict), n_trials=n_trials)

    # Print the best hyperparameters
    print("Best trial:")
    trial = study.best_trial
    print(f"  Value: {trial.value}")
    print("  Params: ")
    for key, value in trial.params.items():
        print(f"    {key}: {value}")

    return study.best_params

# Main function
def main():
    # Load and preprocess data
    print("Loading and preprocessing data...")
    df = pd.read_csv('/content/drive/MyDrive/thesis Final/Implementation/Albania_Dataset.csv')
    data_dict = preprocess_albania_data(df)

    # Run hyperparameter optimization
    print("Starting hyperparameter optimization with Optuna...")
    best_params = run_hyperparameter_optimization(data_dict, n_trials=20)  # Adjust trials as needed

    # Create optimized hyperparameters
    params = HyperParams()

    # Update params with best values from Optuna
    for key, value in best_params.items():
        setattr(params, key, value)

    print("Best hyperparameters:")
    for key, value in best_params.items():
        print(f"    {key}: {value}")

    # Save best hyperparameters to a file
    with open('best_hyperparameters.txt', 'w') as f:
        for key, value in best_params.items():
            f.write(f"{key}: {value}\n")

    # Create data loaders with optimized hyperparameters
    train_loader, val_loader, test_loader = create_dataloaders(data_dict, params)

    # Create model with best hyperparameters
    input_dim = data_dict['train_features'].shape[1]
    output_dim = 1  # Predicting a single value

    model = CNNTransformerLSTM(input_dim, output_dim, params).to(device)

    # Define optimizer and scheduler with best hyperparameters
    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=params.learning_rate,
        weight_decay=params.weight_decay
    )

    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode='min',
        factor=0.5,
        patience=5,
        verbose=True
    )

    # Train the model with best hyperparameters
    print("Training final model with best hyperparameters...")
    num_epochs = 100  # More epochs for final training
    patience = params.patience  # Early stopping patience

    final_model, train_losses, val_losses, _ = train_model(
        model,
        train_loader,
        val_loader,
        optimizer,
        scheduler,
        num_epochs,
        patience,
        device
    )

    # Plot training and validation losses
    plot_losses(train_losses, val_losses)

    # Evaluate on test set
    print("Evaluating model on test set...")
    predictions, targets, metrics = evaluate(final_model, test_loader, data_dict['scaler_y'], device)

    # Print metrics
    print("\nTest Set Metrics:")
    for metric_name, metric_value in metrics.items():
        print(f"{metric_name}: {metric_value:.4f}")

    # Plot predictions vs actual
    print("Plotting predictions vs actual...")
    plot_predictions(targets.flatten(), predictions.flatten(), title='Albania Power Load Forecasting: Predictions vs Actual')

    # Plot a section of the predictions for better visibility
    window_size = 24 * 14  # 14 days assuming hourly data
    if len(targets) > window_size:
        start_idx = len(targets) // 2 - window_size // 2  # Center the window
        end_idx = start_idx + window_size

        # Plot the zoomed section
        plt.figure(figsize=(15, 8))
        plt.plot(targets[start_idx:end_idx], label='Actual', color='blue')
        plt.plot(predictions[start_idx:end_idx], label='Predicted', color='red', alpha=0.7)
        plt.xlabel('Time Step')
        plt.ylabel('Electricity Load (MWh)')
        plt.title('Albania Power Load Forecasting: Predictions vs Actual (2-Week Window)')
        plt.legend()
        plt.grid(True)
        plt.savefig('albania_optuna_zoomed_predictions_plot.png')
        plt.show()

    # Save the model
    torch.save(final_model.state_dict(), 'albania_optuna_model.pth')
    print("Model saved successfully!")

    return final_model, metrics, best_params

if __name__ == "__main__":
    # Try to use seaborn for better visualizations if available
    try:
        import seaborn as sns
        sns.set_theme(style="whitegrid")
    except ImportError:
        print("Seaborn not installed. Using matplotlib for visualizations.")

    # Run the main function
    model, metrics = main()

    print("\nFinal Test Set Metrics:")
    for metric_name, metric_value in metrics.items():
        print(f"{metric_name}: {metric_value:.4f}")
```
